## 1. Baseline

```python
# =============================================================================
# ANOVA BASELINE FEATURE SELECTION (TOP-K = 5,10,15,20)
# =============================================================================

from sklearn.feature_selection import SelectKBest, f_classif
import pandas as pd


# Fit ANOVA once with all features
anova_selector = SelectKBest(
    score_func=f_classif,

    k="all"
)

anova_selector.fit(
    X_train_scaled,
    y_train
)


# Create score dataframe
anova_scores = pd.DataFrame({

    "Feature": X_train_scaled.columns,

    "ANOVA Score": anova_selector.scores_

})


# Sort by ANOVA importance
anova_scores = anova_scores.sort_values(
    by="ANOVA Score",
    ascending=False
).reset_index(drop=True)



# Top-k values
k_values = [5, 10, 15, 20]

anova_results = {}


for k in k_values:

    top_features = (
        anova_scores
        .head(k)["Feature"]
        .tolist()
    )


    # Boolean mask for selecting columns
    mask = X_train_scaled.columns.isin(
        top_features
    )


    X_train_selected = X_train_scaled.loc[
        :, mask
    ]

    X_test_selected = X_test_scaled.loc[
        :, mask
    ]


    anova_results[k] = {

        "features": top_features,

        "X_train": X_train_selected,

        "X_test": X_test_selected,

        "scores": anova_scores.head(k)

    }


    print("="*70)
    print(f"TOP {k} ANOVA FEATURES")
    print("="*70)

    print(top_features)

    print("\nScores:")
    print(anova_scores.head(k))
```

```python
# =============================================================================
# BASELINE MODELS
# =============================================================================

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier


models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        random_state=42,

        eval_metric="logloss",
        scale_pos_weight=(y_train == 0).sum() / (y_train == 1).sum()
    )
}
```

```python
# =============================================================================
# MODEL EVALUATION
# =============================================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

results = []

for k in k_values:

    Xtr = anova_results[k]["X_train"]
    Xts = anova_results[k]["X_test"]

    features = ", ".join(anova_results[k]["features"])

    for model_name, model in models.items():

        model.fit(Xtr, y_train)

        y_pred = model.predict(Xts)

        y_prob = model.predict_proba(Xts)[:,1]

        results.append({

            "Method": "ANOVA",

            "Top-K": k,

            "Model": model_name,

            "Selected Features": features,

            "Accuracy": accuracy_score(y_test, y_pred),

            "Precision": precision_score(y_test, y_pred),

            "Recall": recall_score(y_test, y_pred),

            "F1 Score": f1_score(y_test, y_pred),

            "ROC-AUC": roc_auc_score(y_test, y_prob)

        })
```

```python
results_df = pd.DataFrame(results)

results_df
```

```python
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import f_classif

from doda import DODASelector

from doda.adapters import SklearnAdapter

from doda.knowledge.providers import JSONProvider

from doda.fusion import HadamardFusion
```

```python
# =============================================================================
# ANOVA + DODA FEATURE SELECTION
# =============================================================================

import pandas as pd
from sklearn.feature_selection import SelectKBest, f_classif

# -------------------------------------------------------------------------
# Top-K values
# -------------------------------------------------------------------------

k_values = [5, 10, 15, 20]


# -------------------------------------------------------------------------
# Clinical knowledge provider
# -------------------------------------------------------------------------

provider = JSONProvider(
    "../../../config/clinical_weights/breast_cancer.json"
)


# -------------------------------------------------------------------------
# Fusion method
# -------------------------------------------------------------------------

fusion = HadamardFusion()


# -------------------------------------------------------------------------
# Store DODA results
# -------------------------------------------------------------------------

doda_results = {}


# =============================================================================
# RUN DODA FOR EACH TOP-K
# =============================================================================

for k in k_values:

    print("\n" + "=" * 70)
    print(f"ANOVA + DODA \u2014 TOP {k} FEATURES")
    print("=" * 70)


    # ---------------------------------------------------------------------
    # ANOVA operator
    # ---------------------------------------------------------------------

    anova = SklearnAdapter(
        SelectKBest(
            score_func=f_classif,
            k=k
        )
    )


    # ---------------------------------------------------------------------
    # DODA selector
    # ---------------------------------------------------------------------

    selector = DODASelector(
        operators=[anova],
        provider=provider,
        fusion=fusion,
        top_k=k
    )


    # ---------------------------------------------------------------------
    # Fit DODA on TRAINING data
    # ---------------------------------------------------------------------

    X_train_selected = selector.fit_transform(
        X_train_scaled,
        y_train
    )


    # ---------------------------------------------------------------------
    # Transform TEST data
    # ---------------------------------------------------------------------

    X_test_selected = selector.transform(
        X_test_scaled
    )


    # ---------------------------------------------------------------------
    # Selected features
    # ---------------------------------------------------------------------

    selected_features = selector.get_selected_features()


    print("\nSelected Features:")
    print(selected_features)


    # ---------------------------------------------------------------------
    # Store everything needed for later evaluation
    # ---------------------------------------------------------------------

    doda_results[k] = {

        "selector": selector,

        "X_train": X_train_selected,

        "X_test": X_test_selected,

        "features": selected_features,

        "raw_math_scores": selector.raw_math_scores_,

        "math_scores": selector.math_scores_,

        "clinical_weights": selector.clinical_weights_,

        "final_scores": selector.final_scores_
    }


print("\n" + "=" * 70)
print("DODA FEATURE SELECTION COMPLETED")
print("=" * 70)
```

```python
# CELL 2
# =============================================================================
# INSPECT MATHEMATICAL SCORES
# =============================================================================

k = 10

selector = doda_results[k]["selector"]

# -----------------------------------------------------------------------------
# Raw mathematical scores
# -----------------------------------------------------------------------------

raw_math_scores = pd.DataFrame(
    selector.raw_math_scores_.items(),
    columns=[
        "Feature",
        "Raw Math Score"
    ]
)

# -----------------------------------------------------------------------------
# Normalized mathematical scores
# -----------------------------------------------------------------------------

normalized_math_scores = pd.DataFrame(
    selector.math_scores_.items(),
    columns=[
        "Feature",
        "Normalized Math Score"
    ]
)

# -----------------------------------------------------------------------------
# Combine both
# -----------------------------------------------------------------------------

math_scores = raw_math_scores.merge(
    normalized_math_scores,
    on="Feature"
)

math_scores = math_scores.sort_values(
    by="Normalized Math Score",
    ascending=False
)

print("=" * 70)
print("MATHEMATICAL SCORES")
print("=" * 70)

display(math_scores)
```

```python
# CELL 3
# =============================================================================
# INSPECT CLINICAL WEIGHTS
# =============================================================================

clinical_weights = pd.DataFrame(
    selector.clinical_weights_.items(),
    columns=[
        "Feature",
        "Clinical Weight"
    ]
)

clinical_weights = clinical_weights.sort_values(
    by="Clinical Weight",
    ascending=False
)

print("=" * 70)
print("CLINICAL WEIGHTS")
print("=" * 70)

display(clinical_weights)
```

```python
# CELL 4
# =============================================================================
# INSPECT FINAL DODA SCORES
# =============================================================================

final_scores = pd.DataFrame(
    selector.final_scores_.items(),
    columns=[
        "Feature",
        "Final DODA Score"
    ]
)

final_scores = final_scores.sort_values(
    by="Final DODA Score",
    ascending=False
)

print("=" * 70)
print("FINAL DODA SCORES")
print("=" * 70)

display(final_scores)
```

# 3. ANOVA + DODA: Model Evaluation & Interpretation

## Overall Predictive Performance

The ANOVA baseline produced strong performance across all Top-K settings and all three classifiers. ROC-AUC was consistently high, ranging from 0.984458 to 0.995370, indicating that the statistically selected features carried substantial discriminatory information for the breast cancer classification task. The strongest single ANOVA result by accuracy and F1-score was XGBoost at Top-K = 10, with accuracy = 0.973684, recall = 0.986111, F1-score = 0.979310, and ROC-AUC = 0.992394. The highest ANOVA ROC-AUC was obtained by XGBoost at Top-K = 20 with ROC-AUC = 0.995370.

DODA changed performance in a model- and Top-K-dependent way rather than producing a uniform increase or decrease. At Top-K = 5, DODA improved Logistic Regression accuracy from 0.929825 to 0.938596, recall from 0.902778 to 0.916667, and F1-score from 0.942029 to 0.949640, while ROC-AUC changed only slightly from 0.991071 to 0.990741. Random Forest also improved at Top-K = 5 in accuracy, precision, and F1-score, although ROC-AUC decreased from 0.988261 to 0.982804. In contrast, XGBoost at Top-K = 5 was reduced by DODA across accuracy, precision, recall, F1-score, and ROC-AUC.

At Top-K = 10, DODA maintained exactly the same Logistic Regression performance as ANOVA, but reduced Random Forest and XGBoost performance slightly. For Random Forest, accuracy decreased from 0.956140 to 0.947368 and F1-score from 0.965517 to 0.958333. For XGBoost, accuracy decreased from 0.973684 to 0.964912 and F1-score from 0.979310 to 0.972603, while recall remained unchanged at 0.986111.

At Top-K = 15, DODA maintained the Logistic Regression result exactly and maintained the XGBoost classification metrics, with XGBoost ROC-AUC increasing slightly from 0.992394 to 0.992725. The clearest DODA improvement at this feature size was for Random Forest: accuracy increased from 0.956140 to 0.964912, recall from 0.958333 to 0.972222, and F1-score from 0.965035 to 0.972222, although ROC-AUC decreased from 0.991733 to 0.989087.

At Top-K = 20, DODA again maintained Logistic Regression exactly. Random Forest remained comparable overall, with the same accuracy, improved recall and ROC-AUC, but lower precision. XGBoost improved at Top-K = 20, with accuracy increasing from 0.938596 to 0.947368, recall from 0.944444 to 0.958333, F1-score from 0.951049 to 0.958333, and ROC-AUC from 0.995370 to 0.996032.

Overall, DODA improved performance most clearly for Logistic Regression and Random Forest at Top-K = 5, Random Forest at Top-K = 15, and XGBoost at Top-K = 20. It maintained comparable performance for Logistic Regression at Top-K = 10, 15, and 20, and for XGBoost at Top-K = 15. It reduced performance for XGBoost at Top-K = 5 and 10, and for Random Forest at Top-K = 10. These results indicate that DODA did not uniformly outperform ANOVA, but it often preserved comparable predictive performance while altering the feature ranking.

## Effect of DODA on Feature Selection

The ANOVA ranking was led by `worst concave points`, `worst perimeter`, `mean concave points`, `worst radius`, `mean perimeter`, `worst area`, `mean radius`, `mean area`, `mean concavity`, and `worst concavity`. This shows that the baseline statistical selection was already dominated by lesion size, boundary, and concavity-related measurements.

The DODA ranking retained the same broad feature family but reordered several features after applying the configured clinical weights. The DODA Top-10 ranking was `worst concave points`, `worst perimeter`, `worst radius`, `mean concave points`, `worst area`, `mean perimeter`, `mean radius`, `mean area`, `mean concavity`, and `worst concavity`. Thus, the Top-10 feature set remained the same as ANOVA, but the internal ordering changed.

Several high-ranking features were retained by both approaches across the evaluated Top-K values, including `worst concave points`, `worst perimeter`, `mean concave points`, `worst radius`, `mean perimeter`, `worst area`, `mean radius`, `mean area`, `mean concavity`, and `worst concavity`. These retained features also have high clinical weights in the breast cancer knowledge file: `worst concave points`, `worst perimeter`, `worst radius`, `worst area`, and `worst concavity` each have weight 1.0, while `mean concave points`, `mean perimeter`, `mean radius`, `mean area`, and `mean concavity` each have weight 0.9.

DODA promoted clinically higher-weighted features within the ranking. `worst radius` moved from ANOVA rank 4 to DODA rank 3, and `worst area` moved from ANOVA rank 6 to DODA rank 5, entering the Top-5 DODA subset. This replaced `mean perimeter`, which was rank 5 under ANOVA but rank 6 under DODA. At Top-K = 15, `worst compactness` moved above `mean compactness`, consistent with their configured clinical weights of 0.8 and 0.7 respectively. At Top-K = 20, `mean texture` moved above `concave points error`; their clinical weights are 0.5 and 0.4 respectively.

DODA also demoted some features relative to their ANOVA positions. `mean concave points` moved from rank 3 to rank 4, `mean perimeter` from rank 5 to rank 6, `mean compactness` from rank 11 to rank 12, and `concave points error` from rank 19 to rank 20. These changes do not remove the clinically relevant core of the ANOVA ranking, but they show how DODA shifts preference toward features with stronger configured clinical support when statistical scores are close enough for the weighting to affect order.

The main feature-selection effect in this notebook is therefore refinement rather than wholesale replacement. DODA did not introduce a completely different feature family; instead, it recovered or moved upward clinically weighted features such as `worst area`, `worst radius`, and `worst compactness`. This aligns with the purpose of DODA: to combine statistical evidence with clinical knowledge so that clinically important features are not ranked lower solely because of the mathematical selection score.

## Current Conclusion

The ANOVA baseline demonstrates that a compact subset of statistically selected breast cancer features can achieve strong predictive performance on this dataset. The strongest ANOVA features are concentrated around worst and mean measurements of concave points, perimeter, radius, area, and concavity, all of which are clinically weighted highly in the configured knowledge file.

DODA changes the selected feature space mainly by reordering features within this same clinically meaningful group. Its most visible effect is to promote features with higher clinical weights, including `worst radius`, `worst area`, and `worst compactness`, while demoting lower-weighted or comparatively less clinically prioritised alternatives in some Top-K settings.

Predictive performance under DODA is broadly comparable to the ANOVA baseline, but not uniformly superior. DODA improves some model and Top-K combinations, maintains others, and reduces others. This is important because the results suggest that clinical weighting can refine the selected feature ranking without causing a general collapse in predictive performance, but they do not show consistent performance gains across all classifiers.

For this breast cancer experiment, the results support the idea that clinical knowledge can refine statistically selected features by altering the rank order toward clinically important variables while often maintaining similar predictive performance. However, this single dataset cannot establish that DODA generalizes across diseases, datasets, or cohorts. It also cannot establish statistical significance, because no significance testing was performed in the notebook. The current conclusion is therefore limited to this breast cancer experiment: DODA provides a clinically informed refinement of the ANOVA-selected feature space, with mostly comparable and occasionally improved predictive performance, but further datasets and formal validation are needed before making broader claims.
