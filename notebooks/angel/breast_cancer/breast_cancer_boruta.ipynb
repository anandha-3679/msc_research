```python
# =============================================================================
# BORUTA BASELINE FEATURE SELECTION (TOP-K = 5,10,15,20)
# =============================================================================

from boruta import BorutaPy
from sklearn.ensemble import RandomForestClassifier
import pandas as pd
import numpy as np


# =============================================================================
# BORUTA FEATURE SELECTION
# =============================================================================

# Random Forest used internally by Boruta
rf_boruta = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)


# Initialize Boruta
boruta_selector = BorutaPy(
    estimator=rf_boruta,

    n_estimators="auto",

    verbose=0,

    random_state=42,

    max_iter=100,

    perc=100
)


# -------------------------------------------------------------------------
# Fit Boruta ONLY on training data
# -------------------------------------------------------------------------

boruta_selector.fit(
    X_train_scaled.values,
    y_train.values
)


# =============================================================================
# CREATE BORUTA RANKING
# =============================================================================

boruta_scores = pd.DataFrame({

    "Feature": X_train_scaled.columns,

    "Boruta Rank": boruta_selector.ranking_,

    "Confirmed": boruta_selector.support_,

    "Tentative": boruta_selector.support_weak_

})


# -------------------------------------------------------------------------
# Sort by Boruta rank
# -------------------------------------------------------------------------

boruta_scores = boruta_scores.sort_values(
    by=["Boruta Rank", "Feature"]
).reset_index(drop=True)


print("=" * 70)
print("BORUTA FEATURE RANKING")
print("=" * 70)

print(boruta_scores)


# =============================================================================
# TOP-K FEATURE SELECTION
# =============================================================================

k_values = [5, 10, 15, 20]

boruta_results = {}


for k in k_values:

    # ---------------------------------------------------------------------
    # Select top-k features according to Boruta ranking
    # ---------------------------------------------------------------------

    top_features = (
        boruta_scores
        .head(k)["Feature"]
        .tolist()
    )


    # ---------------------------------------------------------------------
    # Boolean mask for selecting columns
    # ---------------------------------------------------------------------

    mask = X_train_scaled.columns.isin(
        top_features
    )


    # ---------------------------------------------------------------------
    # Select training and testing features
    # ---------------------------------------------------------------------

    X_train_selected = X_train_scaled.loc[
        :, mask
    ]

    X_test_selected = X_test_scaled.loc[
        :, mask
    ]


    # ---------------------------------------------------------------------
    # Store results
    # ---------------------------------------------------------------------

    boruta_results[k] = {

        "features": top_features,

        "X_train": X_train_selected,

        "X_test": X_test_selected,

        "scores": boruta_scores.head(k)

    }


    # ---------------------------------------------------------------------
    # Display
    # ---------------------------------------------------------------------

    print("=" * 70)
    print(f"TOP {k} BORUTA FEATURES")
    print("=" * 70)

    print(top_features)

    print("\nBoruta Ranking:")
    print(
        boruta_scores.head(k)
    )
```

```python
# =============================================================================
# BASELINE MODELS
# =============================================================================

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier


models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        random_state=42,

        eval_metric="logloss",
        scale_pos_weight=(y_train == 0).sum() / (y_train == 1).sum()
    )
}
```

```python
# =============================================================================
# MODEL EVALUATION
# =============================================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)


results = []


for k in k_values:

    Xtr = boruta_results[k]["X_train"]

    Xts = boruta_results[k]["X_test"]

    features = ", ".join(
        boruta_results[k]["features"]
    )


    # ---------------------------------------------------------------------
    # Evaluate every model
    # ---------------------------------------------------------------------

    for model_name, model in models.items():

        # Fit model
        model.fit(
            Xtr,
            y_train
        )


        # Predictions
        y_pred = model.predict(
            Xts
        )


        # Probability predictions
        y_prob = model.predict_proba(
            Xts
        )[:, 1]


        # Store metrics
        results.append({

            "Method": "Boruta",

            "Top-K": k,

            "Model": model_name,

            "Selected Features": features,

            "Accuracy": accuracy_score(
                y_test,
                y_pred
            ),

            "Precision": precision_score(
                y_test,
                y_pred,
                zero_division=0
            ),

            "Recall": recall_score(
                y_test,
                y_pred,
                zero_division=0
            ),

            "F1 Score": f1_score(
                y_test,
                y_pred,
                zero_division=0
            ),

            "ROC-AUC": roc_auc_score(
                y_test,
                y_prob
            )

        })


# =============================================================================
# FINAL BORUTA RESULTS TABLE
# =============================================================================

boruta_results_df = pd.DataFrame(
    results
)


display(
    boruta_results_df
)
```

```python
boruta_results_df.to_csv(
    "../../../results/breast_cancer/baseline_results/boruta_baseline_results.csv",
    index=False
)
```

```python
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import f_classif

from doda import DODASelector

from doda.adapters import SklearnAdapter

from doda.knowledge.providers import JSONProvider

from doda.fusion import HadamardFusion
```

```python
# =============================================================================
# BORUTA + DODA FEATURE SELECTION
# =============================================================================

import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from boruta import BorutaPy


# -------------------------------------------------------------------------
# Top-K values
# -------------------------------------------------------------------------

k_values = [5, 10, 15, 20]


# -------------------------------------------------------------------------
# Clinical knowledge provider
# -------------------------------------------------------------------------

provider = JSONProvider(
    "../../../config/clinical_weights/breast_cancer.json"
)


# -------------------------------------------------------------------------
# Fusion method
# -------------------------------------------------------------------------

fusion = HadamardFusion()


# -------------------------------------------------------------------------
# Store DODA results
# -------------------------------------------------------------------------

doda_results = {}


# =============================================================================
# RUN BORUTA + DODA FOR EACH TOP-K
# =============================================================================

for k in k_values:

    print("\n" + "=" * 70)
    print(f"BORUTA + DODA \u2014 TOP {k} FEATURES")
    print("=" * 70)


    # ---------------------------------------------------------------------
    # Random Forest estimator used by Boruta
    # ---------------------------------------------------------------------

    rf_boruta = RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )


    # ---------------------------------------------------------------------
    # Boruta operator
    # ---------------------------------------------------------------------

    boruta = BorutaPy(
        estimator=rf_boruta,
        n_estimators="auto",
        max_iter=100,
        perc=100,
        random_state=42,
        verbose=0
    )


    # ---------------------------------------------------------------------
    # Wrap Boruta as a DODA operator
    # ---------------------------------------------------------------------

    boruta_operator = SklearnAdapter(
        boruta
    )


    # ---------------------------------------------------------------------
    # DODA selector
    # ---------------------------------------------------------------------

    selector = DODASelector(
        operators=[boruta_operator],
        provider=provider,
        fusion=fusion,
        top_k=k
    )


    # ---------------------------------------------------------------------
    # Fit DODA on TRAINING data
    # ---------------------------------------------------------------------

    X_train_selected = selector.fit_transform(
        X_train_scaled,
        y_train
    )


    # ---------------------------------------------------------------------
    # Transform TEST data
    # ---------------------------------------------------------------------

    X_test_selected = selector.transform(
        X_test_scaled
    )


    # ---------------------------------------------------------------------
    # Selected features
    # ---------------------------------------------------------------------

    selected_features = selector.get_selected_features()


    print("\nSelected Features:")
    print(selected_features)


    # ---------------------------------------------------------------------
    # Store everything needed for later evaluation
    # ---------------------------------------------------------------------

    doda_results[k] = {

        "selector": selector,

        "X_train": X_train_selected,

        "X_test": X_test_selected,

        "features": selected_features,

        "raw_math_scores": selector.raw_math_scores_,

        "math_scores": selector.math_scores_,

        "clinical_weights": selector.clinical_weights_,

        "final_scores": selector.final_scores_
    }


print("\n" + "=" * 70)
print("BORUTA + DODA FEATURE SELECTION COMPLETED")
print("=" * 70)
```

# 3. Boruta + DODA: Experiment Interpretation

## 3.1 Overview

This experiment compares a **data-driven Boruta baseline** with **Boruta + DODA**, where configured clinical knowledge is used to refine feature prioritization. The comparison evaluates whether DODA changes the Boruta-derived ranking and whether those changes improve, reduce, or maintain predictive performance across Top-5, Top-10, Top-15, and Top-20 feature subsets using Logistic Regression, Random Forest, and XGBoost.

The results should therefore be interpreted at two levels: **changes in feature prioritization** and **changes in downstream predictive performance**

## 3.2 Boruta Baseline Results

The Boruta baseline produced strong predictive performance across all Top-K feature subsets. Performance was not strictly improved by adding more features, indicating that the optimal subset size depended on the downstream model.

| Model | Strongest Boruta Top-K | Accuracy | Recall | F1-score | ROC-AUC |
|---|---:|---:|---:|---:|---:|
| Logistic Regression | Top-20 | 0.9561 | 0.9444 | 0.9645 | 0.9970 |
| Random Forest | Top-10 / Top-15 | 0.9561 | 0.9583 | 0.9650 | 0.9942–0.9899 |
| XGBoost | Top-15 | 0.9649 | 0.9861 | 0.9726 | 0.9940 |

**Observed patterns:**
- Logistic Regression reached its highest Accuracy and ROC-AUC with Top-20.
- Random Forest performed best in Accuracy with Top-10 and Top-15, while additional features did not improve Accuracy further.
- XGBoost achieved the strongest baseline Accuracy, Recall, and F1-score at Top-15.
- Top-10 to Top-20 generally produced stronger results than Top-5, although the best subset size remained model-dependent.

## 3.3 Boruta + DODA Results

After applying DODA, several configurations improved relative to the Boruta baseline, while others were reduced or remained comparable.

| Strong DODA Configuration | Accuracy | Recall | F1-score | ROC-AUC |
|---|---:|---:|---:|---:|
| Top-10 XGBoost | 0.9737 | 0.9861 | 0.9793 | 0.9921 |
| Top-20 Logistic Regression | 0.9561 | 0.9444 | 0.9645 | 0.9944 |
| Top-20 XGBoost | 0.9474 | 0.9722 | 0.9589 | 0.9967 |

**Observed results:**
- The clearest Accuracy and F1 improvement occurred for **XGBoost at Top-10**.
- Logistic Regression improved in several smaller feature subsets but did not exceed its baseline Accuracy at Top-20.
- Random Forest showed mixed results, with some reductions in Accuracy and ROC-AUC.
- Several configurations remained numerically close to the Boruta baseline, particularly at larger Top-K values.

Therefore, DODA did not produce a uniform improvement across all models and feature-set sizes.

## 3.4 Effect of DODA on Feature Ranking

Boruta assigned Rank 1 to a large group of confirmed features, including several mean, error, and worst measurements. This indicates limited differentiation among many of the Boruta-confirmed features because they shared the same top rank.

DODA substantially reprioritized the feature order by promoting clinically weighted **worst** measurements to the top of the ranking.

| Feature | Boruta Position/Status | DODA Position |
|---|---|---:|
| worst radius | Boruta Rank 1 | 1 |
| worst perimeter | Boruta Rank 1 | 2 |
| worst area | Boruta Rank 1 | 3 |
| worst concavity | Boruta Rank 1 | 4 |
| worst concave points | Boruta Rank 1 | 5 |
| mean radius | Boruta Rank 1 | 6 |
| mean perimeter | Boruta Rank 1 | 7 |
| mean area | Boruta Rank 1 | 8 |
| mean concavity | Boruta Rank 1 | 9 |
| mean concave points | Boruta Rank 1 | 10 |

**Interpretation:**
- DODA promoted clinically weighted features such as worst radius, worst perimeter, worst area, worst concavity, and worst concave points.
- These features were already Boruta-confirmed, so their promotion should be interpreted as **clinical reprioritization**, not as evidence that DODA statistically proved them more important.
- The main effect of DODA was a substantial change in ranking order and Top-K membership.
- Boruta's large group of Rank-1 features provided limited internal differentiation, while DODA imposed an additional prioritization layer based on configured clinical knowledge.

## 3.5 Top-K Feature Selection Comparison

### Top-5

**Boruta Top-5:**
- area error
- concavity error
- mean area
- mean compactness
- mean concave points

**Boruta + DODA Top-5:**
- worst radius
- worst perimeter
- worst area
- worst concavity
- worst concave points

The Top-5 subsets are completely different in membership and clearly show the effect of DODA's clinical prioritization.

### Top-10

Boruta retained primarily mean and error measurements, while DODA prioritized five worst measurements followed by mean radius, mean perimeter, mean area, mean concavity, and mean concave points.

The Top-10 membership therefore changed substantially rather than simply changing the order of an identical subset.

### Top-15

DODA retained several features also represented among Boruta-confirmed variables, but the ordering shifted toward clinically prioritized worst measurements and selected mean measurements.

### Top-20

At Top-20, both approaches covered a broader feature space, but the ranking order and selected subset membership still differed.

### Overall Interpretation

The notebook output shows `Top-K Same: False` for Top-5, Top-10, Top-15, and Top-20. Therefore, DODA did not merely reorder an identical Top-K subset. Instead, it changed feature membership at every evaluated Top-K level.

Feature-set stability increased conceptually as more features were included, but the actual Top-K outputs remained different across all four subset sizes.

## 3.7 Logistic Regression Results

For Logistic Regression, DODA generally improved performance at Top-5, Top-10, and Top-15.

- **Best Boruta configuration:** Top-20 with Accuracy = 0.9561, F1-score = 0.9645, and ROC-AUC = 0.9970.
- **Best DODA Accuracy:** Top-20 with Accuracy = 0.9561.
- **Largest DODA improvement:** Top-10, where Accuracy increased from 0.8947 to 0.9123 and F1-score increased from 0.9118 to 0.9275.
- At Top-20, Accuracy, Recall, and F1-score were identical between Boruta and DODA, although ROC-AUC was slightly lower with DODA.

Overall, DODA improved smaller and medium-sized feature subsets for Logistic Regression, while the largest subset preserved essentially the same predictive performance.

## 3.8 Random Forest Results

Random Forest showed a mixed response to DODA-adjusted feature subsets.

- **Best Boruta Accuracy:** Top-10 and Top-15, both at 0.9561.
- **Best DODA Accuracy:** Top-20 at 0.9474.
- Top-5, Top-10, and Top-15 showed lower Accuracy and F1-score after DODA.
- Top-20 improved Accuracy from 0.9386 to 0.9474 and Recall from 0.9306 to 0.9583, while ROC-AUC remained identical at 0.9939.

These results indicate that Random Forest did not consistently benefit from the DODA-adjusted ranking. Larger Top-K values produced the strongest DODA result, but still did not exceed the strongest Boruta Accuracy.

## 3.9 XGBoost Results

XGBoost showed the strongest improvement from DODA at Top-10.

- **Best Boruta configuration:** Top-15 with Accuracy = 0.9649, Recall = 0.9861, F1-score = 0.9726, and ROC-AUC = 0.9940.
- **Best DODA Accuracy, Recall, and F1-score:** Top-10 with Accuracy = 0.9737, Recall = 0.9861, and F1-score = 0.9793.
- Top-5 also improved in Accuracy, Recall, and F1-score, although ROC-AUC was slightly lower.
- Top-15 showed a substantial reduction after DODA.
- Top-20 showed slightly lower Accuracy and F1-score but achieved the highest DODA ROC-AUC of 0.9967.

Therefore, XGBoost benefited from DODA at smaller feature subsets, particularly Top-10, but the effect depended strongly on feature-set size.

## 3.11 Importance of Accuracy, Recall, F1-score, and ROC-AUC

The four metrics provide different perspectives on model performance:

- **Accuracy** measures overall classification correctness.
- **Recall** measures the ability to correctly identify positive cases under the experiment's target encoding.
- **F1-score** balances Precision and Recall.
- **ROC-AUC** measures the model's ability to discriminate between classes across decision thresholds.

In this experiment, the configuration with the highest Accuracy and F1-score, Boruta + DODA with Top-10 XGBoost, was not the same configuration with the highest ROC-AUC, Boruta with Top-20 Logistic Regression.

This demonstrates why performance should not be judged using Accuracy alone. Different feature-selection and model combinations may optimize different aspects of predictive performance.

## 3.12 Effect of Clinical Knowledge

DODA contributed a clinically informed prioritization layer to the Boruta-derived feature ranking.

The observed results show that:
- Clinically high-weighted worst measurements moved to the top of the DODA ranking.
- Several features already supported by Boruta remained important, but their ordering and Top-K membership changed.
- The resulting feature subsets differed from the Boruta Top-K subsets at all evaluated values of K.
- Changes in feature prioritization did not automatically translate into improved predictive performance.
- The effect depended on both the downstream model and the number of selected features.

The strongest observed benefit occurred for **XGBoost at Top-10**, while other configurations remained comparable or showed reduced performance. Therefore, clinical prioritization and predictive optimization were related in this experiment but were not equivalent objectives.

## 3.13 Important Observations

- DODA had a **model-dependent effect**: XGBoost showed the strongest improvement, while Random Forest showed several reductions.
- The effect was also **Top-K-dependent**: the same model could improve at one subset size and decrease at another.
- DODA substantially changed the feature ranking and Top-K membership, but these ranking changes did not always produce equally large changes in predictive performance.
- Logistic Regression generally benefited from DODA at smaller and medium-sized subsets, while its Top-20 performance was essentially preserved.
- Random Forest did not consistently benefit from DODA and performed best with the larger DODA subset at Top-20.
- XGBoost achieved the strongest overall Accuracy and F1-score with DODA at Top-10.
- Some configurations showed meaningful changes in feature selection while predictive performance remained relatively comparable.
- The experiment demonstrates that the same feature-ranking modification can produce different responses across classifiers.

## 3.14 Main Findings

1. Boruta provided a strong data-driven baseline, with many confirmed features sharing Rank 1 and producing high predictive performance across all three classifiers.
2. DODA substantially refined the ranking by promoting clinically weighted worst measurements and changing Top-K feature membership.
3. Important data-driven features were not simply discarded; several Boruta-confirmed features remained prominent after DODA, although their priority changed.
4. Predictive performance was generally stable and remained high across both approaches, but the effect of DODA varied by classifier and Top-K size.
5. DODA did **not** consistently improve performance across all configurations.
6. The strongest observed benefit was Boruta + DODA with Top-10 features and XGBoost, which achieved the highest Accuracy and F1-score in the experiment.

## 3.15 Research Interpretation

> Does incorporating clinical knowledge into feature-selection rankings consistently improve or preserve predictive performance across datasets and selection strategies?

This Boruta experiment provides **one dataset-specific piece of evidence** for this broader question.

The results suggest that incorporating configured clinical knowledge can substantially alter feature prioritization while maintaining or, in some configurations, improving predictive performance. However, the effect was not consistent across all models or Top-K subsets.

In this experiment:
- The breast cancer dataset represents only one predictive setting.
- Boruta represents only one underlying feature-selection strategy.
- DODA's predictive effect depended on the downstream classifier.
- Feature-set size influenced whether performance improved, remained comparable, or decreased.
- Clinical prioritization and purely predictive optimization were related but not identical.

Therefore, this experiment supports further investigation rather than establishing a general rule that clinically informed feature ranking consistently improves predictive performance.

## 3.16 Important Limitations

- The experiment was conducted using a single breast cancer dataset.
- The findings are specific to the current train-test split, model settings, evaluation procedure, and Top-K values.
- No statistical significance testing was performed in the reported results.
- The experiment does not validate the approach across multiple diseases or independent datasets.
- Clinical weights are configured inputs, meaning their specification can directly influence the resulting DODA ranking.
- The results therefore cannot establish universal generalization or universal superiority of DODA.

## 3.17 Final Conclusion

In this breast cancer experiment, Boruta provided a strong data-driven baseline for feature selection, producing high predictive performance across Logistic Regression, Random Forest, and XGBoost.

DODA substantially refined the Boruta-derived prioritization by promoting clinically weighted features and changing Top-K feature membership at every evaluated subset size. However, these ranking changes did not produce a uniform predictive advantage.

The strongest observed improvement occurred with **Boruta + DODA using Top-10 features and XGBoost**, which achieved the highest Accuracy and F1-score. Other configurations remained comparable to the Boruta baseline, while some showed reduced performance.

Overall, the results suggest that combining data-driven feature selection with configured clinical knowledge can be useful as a feature-prioritization strategy, but its predictive effect depends on the classifier and the number of selected features. This experiment therefore provides dataset-specific evidence rather than demonstrating that DODA is universally superior, and broader validation across datasets and feature-selection strategies is required.
