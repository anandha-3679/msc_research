```python
# =============================================================================
# MRMR BASELINE FEATURE SELECTION
# TOP-K = 5, 10, 15, 20
# =============================================================================

from mrmr import mrmr_classif
import pandas as pd


# -------------------------------------------------------------------------
# Align indices
# -------------------------------------------------------------------------

X_train_mrmr = X_train_scaled.copy()
y_train_mrmr = y_train.copy()

X_train_mrmr = X_train_mrmr.reset_index(drop=True)
y_train_mrmr = y_train_mrmr.reset_index(drop=True)


# -------------------------------------------------------------------------
# Top-K values
# -------------------------------------------------------------------------

k_values = [5, 10, 15, 20]


# -------------------------------------------------------------------------
# Store MRMR results
# -------------------------------------------------------------------------

mrmr_results = {}


# =============================================================================
# RUN MRMR
# =============================================================================

for k in k_values:

    print("\n" + "=" * 70)
    print(f"TOP {k} MRMR FEATURES")
    print("=" * 70)


    # ---------------------------------------------------------------------
    # MRMR feature selection
    # ---------------------------------------------------------------------

    selected_features = mrmr_classif(
        X=X_train_mrmr,
        y=y_train_mrmr,
        K=k
    )


    # ---------------------------------------------------------------------
    # Select features
    # ---------------------------------------------------------------------

    X_train_selected = X_train_scaled[
        selected_features
    ]

    X_test_selected = X_test_scaled[
        selected_features
    ]


    # ---------------------------------------------------------------------
    # Store results
    # ---------------------------------------------------------------------

    mrmr_results[k] = {

        "features": selected_features,

        "X_train": X_train_selected,

        "X_test": X_test_selected

    }


    # ---------------------------------------------------------------------
    # Display
    # ---------------------------------------------------------------------

    print("\nSelected Features:")

    print(selected_features)


print("\n" + "=" * 70)
print("MRMR FEATURE SELECTION COMPLETED")
print("=" * 70)
```

```python
# =============================================================================
# BASELINE MODELS
# =============================================================================

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier


models = {

    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        random_state=42,
        eval_metric="logloss",
        scale_pos_weight=(
            (y_train == 0).sum() /
            (y_train == 1).sum()
        )
    )
}
```

```python
# =============================================================================
# MRMR MODEL EVALUATION
# =============================================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)


mrmr_model_results = []


for k in k_values:

    Xtr = mrmr_results[k]["X_train"]

    Xts = mrmr_results[k]["X_test"]

    features = ", ".join(
        mrmr_results[k]["features"]
    )


    for model_name, model in models.items():

        # -------------------------------------------------------------
        # Train
        # -------------------------------------------------------------

        model.fit(
            Xtr,
            y_train
        )


        # -------------------------------------------------------------
        # Predictions
        # -------------------------------------------------------------

        y_pred = model.predict(
            Xts
        )


        y_prob = model.predict_proba(
            Xts
        )[:, 1]


        # -------------------------------------------------------------
        # Store metrics
        # -------------------------------------------------------------

        mrmr_model_results.append({

            "Method": "MRMR",

            "Top-K": k,

            "Model": model_name,

            "Selected Features": features,

            "Accuracy": accuracy_score(
                y_test,
                y_pred
            ),

            "Precision": precision_score(
                y_test,
                y_pred,
                zero_division=0
            ),

            "Recall": recall_score(
                y_test,
                y_pred,
                zero_division=0
            ),

            "F1 Score": f1_score(
                y_test,
                y_pred,
                zero_division=0
            ),

            "ROC-AUC": roc_auc_score(
                y_test,
                y_prob
            )
        })


# =============================================================================
# CREATE RESULTS DATAFRAME
# =============================================================================

mrmr_results_df = pd.DataFrame(
    mrmr_model_results
)


display(mrmr_results_df)
```

```python
# =============================================================================
# MRMR ADAPTER FOR DODA
# =============================================================================

import numpy as np


class MRMRAdapter:

    def __init__(self, K):
        self.K = K
        self.importance = {}

    def fit(self, X, y):

        # Run MRMR
        selected_features = mrmr_classif(
            X=X,
            y=y,
            K=self.K
        )

        # -------------------------------------------------------------
        # Convert MRMR ranking into importance scores
        # First feature = highest score
        # -------------------------------------------------------------

        n_features = len(selected_features)

        scores = np.zeros(len(X.columns))

        feature_to_index = {
            feature: i
            for i, feature in enumerate(X.columns)
        }

        for rank, feature in enumerate(selected_features):

            index = feature_to_index[feature]

            scores[index] = (
                n_features - rank
            ) / n_features

        # -------------------------------------------------------------
        # Store importance
        # -------------------------------------------------------------

        self.importance = dict(
            zip(
                X.columns,
                scores
            )
        )

        return self

    def get_importance(self):

        return self.importance

    def get_name(self):

        return "MRMR"
```

```python
# =============================================================================
# MRMR + DODA FEATURE SELECTION
# =============================================================================

import numpy as np
import pandas as pd


# -------------------------------------------------------------------------
# Top-K values
# -------------------------------------------------------------------------

k_values = [5, 10, 15, 20]


# -------------------------------------------------------------------------
# Clinical knowledge provider
# -------------------------------------------------------------------------

provider = JSONProvider(
    "../../../config/clinical_weights/breast_cancer.json"
)


# -------------------------------------------------------------------------
# Fusion method
# -------------------------------------------------------------------------

fusion = HadamardFusion()


# -------------------------------------------------------------------------
# MRMR Adapter
# -------------------------------------------------------------------------

class MRMRAdapter:

    def __init__(self, ranked_features):

        self.ranked_features = ranked_features
        self.importance = {}


    def fit(self, X, y):

        n_features = len(self.ranked_features)

        # Convert MRMR ranking into decreasing scores
        # Rank 1 = highest score
        # Rank n = lowest score

        for rank, feature in enumerate(
            self.ranked_features,
            start=1
        ):

            self.importance[feature] = (
                n_features - rank + 1
            ) / n_features

        return self


    def get_importance(self):

        return self.importance


    def get_name(self):

        return "MRMR"


# -------------------------------------------------------------------------
# Store results
# -------------------------------------------------------------------------

mrmr_doda_results = {}


# =============================================================================
# RUN MRMR + DODA
# =============================================================================

for k in k_values:

    print("\n" + "=" * 70)
    print(f"MRMR + DODA \u2014 TOP {k} FEATURES")
    print("=" * 70)


    # ---------------------------------------------------------------------
    # Create MRMR adapter
    # ---------------------------------------------------------------------

    mrmr_adapter = MRMRAdapter(
        mrmr_ranked_features
    )


    # ---------------------------------------------------------------------
    # DODA selector
    # ---------------------------------------------------------------------

    selector = DODASelector(

        operators=[
            mrmr_adapter
        ],

        provider=provider,

        fusion=fusion,

        top_k=k
    )


    # ---------------------------------------------------------------------
    # Fit DODA on training data
    # ---------------------------------------------------------------------

    X_train_selected = selector.fit_transform(
        X_train_scaled,
        y_train
    )


    # ---------------------------------------------------------------------
    # Transform test data
    # ---------------------------------------------------------------------

    X_test_selected = selector.transform(
        X_test_scaled
    )


    # ---------------------------------------------------------------------
    # Selected features
    # ---------------------------------------------------------------------

    selected_features = (
        selector.get_selected_features()
    )


    print("\nSelected Features:")
    print(selected_features)


    # ---------------------------------------------------------------------
    # Store results
    # ---------------------------------------------------------------------

    mrmr_doda_results[k] = {

        "selector": selector,

        "X_train": X_train_selected,

        "X_test": X_test_selected,

        "features": selected_features,

        "raw_math_scores":
            selector.raw_math_scores_,

        "math_scores":
            selector.math_scores_,

        "clinical_weights":
            selector.clinical_weights_,

        "final_scores":
            selector.final_scores_
    }


print("\n" + "=" * 70)
print("MRMR + DODA FEATURE SELECTION COMPLETED")
print("=" * 70)
```

# 3. mRMR + DODA: Model Evaluation & Interpretation

## 3.1 Overall Performance

Both mRMR and mRMR + DODA produced strong predictive performance across the tested Top-K feature subsets. Most configurations achieved accuracy and F1-scores above 0.92, with ROC-AUC values generally close to or above 0.98.

The best mRMR baseline configuration was **Logistic Regression with Top-K = 20**, which achieved the highest baseline accuracy and F1-score, while XGBoost reached the highest baseline ROC-AUC at Top-K = 15.

The best mRMR + DODA configurations were **XGBoost with Top-K = 10 and Top-K = 20** for accuracy and F1-score, with XGBoost at Top-K = 20 also matching the highest DODA ROC-AUC.

DODA did **not** consistently improve predictive performance. Its effect was configuration-dependent: it improved Logistic Regression at Top-K = 5, maintained several Top-K = 10 and Top-K = 15 results, and reduced performance in some Random Forest and XGBoost settings. The main trade-off was that clinical weighting refined the feature ranking, but this did not always translate into higher model performance.

Important high-ranked features were retained, including `worst concave points`, `worst perimeter`, `mean concave points`, `worst radius`, `mean perimeter`, `worst area`, `mean radius`, `worst concavity`, `mean concavity`, and `mean area`.

DODA mainly refined the ordering rather than substantially changing the Top-K = 10 subset. In the Top-K = 5 subset, `worst radius` and `worst area` moved upward, while `mean perimeter` left the Top-5. At Top-K = 20, `mean symmetry` entered the DODA-selected subset, while `concave points error` was no longer included.

The upward movement of features such as `worst radius`, `worst area`, and `worst concavity` is aligned with their higher configured clinical weights in the executed outputs. However, this should be interpreted as ranking refinement rather than direct evidence of clinical improvement. The final DODA scores also show some repeated zero values among lower-ranked features, so the ordering of those low-scoring features should be interpreted cautiously.


## 3.5 Model-Level Comparison

### Logistic Regression

Logistic Regression improved at Top-K = 5, was maintained at Top-K = 10 and Top-K = 15, and was reduced at Top-K = 20 for accuracy, recall, and F1-score. Its best DODA configuration was Top-K = 20, although the mRMR baseline at Top-K = 20 remained slightly stronger. Overall, DODA had a mostly comparable effect for Logistic Regression, with one improvement at the smallest feature subset.

### Random Forest

Random Forest performance was maintained at Top-K = 5 and Top-K = 10 for accuracy and F1-score, but reduced at Top-K = 15 and Top-K = 20. Its best DODA configuration was Top-K = 10. Overall, DODA did not improve Random Forest performance in the executed results.

### XGBoost

XGBoost was reduced at Top-K = 5 and Top-K = 15, maintained at Top-K = 10, and improved in F1-score at Top-K = 20. Its best DODA configurations were Top-K = 10 and Top-K = 20, both reaching 0.964912 accuracy, with Top-K = 20 achieving the highest DODA ROC-AUC. Overall, DODA had a configuration-dependent effect for XGBoost rather than a consistently positive one.

## 3.7 Important Observations

- Predictive performance was strong for both approaches, with most configurations producing high F1-scores and ROC-AUC values.
- DODA's effect depended on both the model and the Top-K feature subset.
- The Top-K = 10 DODA results were especially stable, with Logistic Regression and XGBoost matching the mRMR baseline and Random Forest maintaining F1-score.
- DODA changed the ranking more clearly than it changed the overall selected feature set, especially for the Top-K = 10 and Top-K = 15 subsets.
- Clinical weighting increased the priority of several high-weight `worst` diagnostic features, but the resulting ranking changes did not guarantee better predictive performance.
- The experiment supports cautious interpretation only; it does not establish that clinical knowledge universally improves mRMR-based classification.


## 3.8 Conclusion

DODA did not consistently improve mRMR predictive performance in this breast cancer experiment. Some configurations improved, particularly Logistic Regression at Top-K = 5 and XGBoost at Top-K = 20, but other configurations were maintained or reduced.

DODA preserved competitive performance in several cases, especially at Top-K = 10, where multiple metrics remained unchanged or closely comparable to the mRMR baseline. This suggests that clinical weighting can be introduced without necessarily degrading performance in some settings.

DODA meaningfully altered the feature ranking by promoting clinically weighted features such as `worst radius`, `worst area`, and `worst concavity`. These changes were aligned with the configured clinical weights shown in the executed outputs. However, the results do not support a claim that these ranking changes directly caused clinical or predictive improvement.

Overall, the evidence suggests that clinical knowledge acted mainly as a refinement of statistically selected mRMR features rather than as a universally performance-improving mechanism. The experiment supports the value of DODA as an interpretable ranking adjustment method, but its predictive benefit remains model- and Top-K-dependent.
