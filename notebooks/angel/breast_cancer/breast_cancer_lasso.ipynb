````python
# =============================================================================
# STEP 1: LOAD PROCESSED DATASET
# =============================================================================

import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")


# Load cleaned dataset
df = pd.read_csv(
    "../../data/processed/breast_cancer_cleaned.csv"
)


print("=" * 70)
print("PROCESSED DATASET")
print("=" * 70)

print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")


display(df.head())
````

## 2. Baseline

````python
# =============================================================================
# LASSO BASELINE FEATURE SELECTION (TOP-K = 5,10,15,20)
# =============================================================================

from sklearn.linear_model import LogisticRegression
import pandas as pd
import numpy as np


# Fit LASSO once with all features
# Logistic Regression with L1 penalty = LASSO-style feature selection

lasso_selector = LogisticRegression(
    penalty="l1",
    solver="liblinear",
    C=0.1,
    max_iter=2000,
    class_weight="balanced",
    random_state=42
)


lasso_selector.fit(
    X_train_scaled,
    y_train
)


# =============================================================================
# Extract LASSO coefficients
# =============================================================================

lasso_coefficients = lasso_selector.coef_[0]


# Use absolute coefficient magnitude as feature importance
lasso_scores = pd.DataFrame({

    "Feature": X_train_scaled.columns,

    "LASSO Coefficient": lasso_coefficients,

    "LASSO Score": np.abs(lasso_coefficients)

})


# Sort by absolute coefficient magnitude
lasso_scores = lasso_scores.sort_values(
    by="LASSO Score",
    ascending=False
).reset_index(drop=True)


# =============================================================================
# Top-K values
# =============================================================================

k_values = [5, 10, 15, 20]

lasso_results = {}


for k in k_values:

    top_features = (
        lasso_scores
        .head(k)["Feature"]
        .tolist()
    )


    # Boolean mask for selecting columns
    mask = X_train_scaled.columns.isin(
        top_features
    )


    X_train_selected = X_train_scaled.loc[
        :, mask
    ]

    X_test_selected = X_test_scaled.loc[
        :, mask
    ]


    lasso_results[k] = {

        "features": top_features,

        "X_train": X_train_selected,

        "X_test": X_test_selected,

        "scores": lasso_scores.head(k)

    }


    print("=" * 70)
    print(f"TOP {k} LASSO FEATURES")
    print("=" * 70)

    print(top_features)

    print("\nScores:")
    print(lasso_scores.head(k))
````

````python
# =============================================================================
# BASELINE MODELS
# =============================================================================

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier


models = {

    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        random_state=42,
        eval_metric="logloss",
        scale_pos_weight=(
            (y_train == 0).sum() /
            (y_train == 1).sum()
        )
    )
}
````

````python
# =============================================================================
# MODEL EVALUATION — LASSO
# =============================================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

results_lasso = []


for k in k_values:

    Xtr = lasso_results[k]["X_train"]

    Xts = lasso_results[k]["X_test"]

    features = ", ".join(
        lasso_results[k]["features"]
    )


    for model_name, model in models.items():

        model.fit(
            Xtr,
            y_train
        )


        y_pred = model.predict(
            Xts
        )


        y_prob = model.predict_proba(
            Xts
        )[:, 1]


        results_lasso.append({

            "Method": "LASSO",

            "Top-K": k,

            "Model": model_name,

            "Selected Features": features,

            "Accuracy": accuracy_score(
                y_test,
                y_pred
            ),

            "Precision": precision_score(
                y_test,
                y_pred
            ),

            "Recall": recall_score(
                y_test,
                y_pred
            ),

            "F1 Score": f1_score(
                y_test,
                y_pred
            ),

            "ROC-AUC": roc_auc_score(
                y_test,
                y_prob
            )

        })


lasso_evaluation = pd.DataFrame(
    results_lasso
)

display(lasso_evaluation)
````

````python
# =============================================================================
# LASSO + DODA FEATURE SELECTION
# =============================================================================

import pandas as pd
from sklearn.linear_model import LogisticRegression


# -------------------------------------------------------------------------
# Top-K values
# -------------------------------------------------------------------------

k_values = [5, 10, 15, 20]


# -------------------------------------------------------------------------
# Clinical knowledge provider
# -------------------------------------------------------------------------

provider = JSONProvider(
    "../../../config/clinical_weights/breast_cancer.json"
)


# -------------------------------------------------------------------------
# Fusion method
# -------------------------------------------------------------------------

fusion = HadamardFusion()


# -------------------------------------------------------------------------
# Store DODA results
# -------------------------------------------------------------------------

doda_results = {}


# =============================================================================
# RUN LASSO + DODA FOR EACH TOP-K
# =============================================================================

for k in k_values:

    print("\n" + "=" * 70)
    print(f"LASSO + DODA \u2014 TOP {k} FEATURES")
    print("=" * 70)


    # ---------------------------------------------------------------------
    # LASSO operator
    # ---------------------------------------------------------------------

    lasso = SklearnAdapter(

        LogisticRegression(
            penalty="l1",
            solver="liblinear",
            C=0.1,
            max_iter=2000,
            class_weight="balanced",
            random_state=42
        )

    )


    # ---------------------------------------------------------------------
    # DODA selector
    # ---------------------------------------------------------------------

    selector = DODASelector(

        operators=[lasso],

        provider=provider,

        fusion=fusion,

        top_k=k
    )


    # ---------------------------------------------------------------------
    # Fit DODA on TRAINING data
    # ---------------------------------------------------------------------

    X_train_selected = selector.fit_transform(

        X_train_scaled,

        y_train

    )


    # ---------------------------------------------------------------------
    # Transform TEST data
    # ---------------------------------------------------------------------

    X_test_selected = selector.transform(

        X_test_scaled

    )


    # ---------------------------------------------------------------------
    # Selected features
    # ---------------------------------------------------------------------

    selected_features = (
        selector.get_selected_features()
    )


    print("\nSelected Features:")
    print(selected_features)


    # ---------------------------------------------------------------------
    # Store everything needed for later evaluation
    # ---------------------------------------------------------------------

    doda_results[k] = {

        "selector": selector,

        "X_train": X_train_selected,

        "X_test": X_test_selected,

        "features": selected_features,

        "raw_math_scores":
            selector.raw_math_scores_,

        "math_scores":
            selector.math_scores_,

        "clinical_weights":
            selector.clinical_weights_,

        "final_scores":
            selector.final_scores_
    }


print("\n" + "=" * 70)
print("LASSO + DODA FEATURE SELECTION COMPLETED")
print("=" * 70)
````

````python
# =============================================================================
# STEP 21: CROSS-VALIDATION CONFIGURATION
# =============================================================================
# LASSO vs LASSO + DODA
#
# The fixed-split experiment above is retained unchanged.
# This section adds fold-wise and repeated-CV validation.
#
# LASSO requires standardized predictors. Therefore, scaling is fitted
# separately inside every CV training fold to prevent data leakage.
#
# DODA continues to use the existing Hadamard fusion method exactly as
# defined in the fixed-split experiment.

import os
import numpy as np
import pandas as pd

from itertools import combinations

from sklearn.model_selection import (
    StratifiedKFold,
    RepeatedStratifiedKFold
)

from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

from doda import DODASelector
from doda.adapters import SklearnAdapter
from doda.knowledge.providers import JSONProvider
from doda.fusion import HadamardFusion


# -------------------------------------------------------------------------
# Research design
# -------------------------------------------------------------------------

TOP_K_VALUES = [5, 10, 15, 20]

N_SPLITS = 5
N_REPEATS = 5
RANDOM_STATE = 42

WEIGHTS_FILE = "../../../config/clinical_weights/breast_cancer.json"

print("=" * 80)
print("BREAST CANCER LASSO CROSS-VALIDATION CONFIGURATION")
print("=" * 80)

print("Top-K values:", TOP_K_VALUES)
print("5-fold CV:", N_SPLITS)
print(
    "Repeated CV:",
    f"{N_SPLITS} folds × {N_REPEATS} repeats"
)
print(
    "Total repeated runs:",
    N_SPLITS * N_REPEATS
)
print("Random state:", RANDOM_STATE)
print("Clinical weights:", WEIGHTS_FILE)


# -------------------------------------------------------------------------
# Fresh LASSO factory
# -------------------------------------------------------------------------

def make_lasso(random_state=RANDOM_STATE):
    """Create a fresh L1-regularized logistic regression selector."""

    return LogisticRegression(
        penalty="l1",
        solver="liblinear",
        C=0.1,
        max_iter=2000,
        class_weight="balanced",
        random_state=random_state
    )


def make_doda_lasso(top_k):
    """Create a fresh LASSO + DODA selector using Hadamard fusion."""

    lasso_operator = SklearnAdapter(
        make_lasso(RANDOM_STATE)
    )

    provider = JSONProvider(
        WEIGHTS_FILE
    )

    return DODASelector(
        operators=[lasso_operator],
        provider=provider,
        fusion=HadamardFusion(),
        top_k=top_k
    )


# -------------------------------------------------------------------------
# Downstream models
# -------------------------------------------------------------------------

cv_models = {

    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=RANDOM_STATE
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),

    "XGBoost": XGBClassifier(
        random_state=RANDOM_STATE,
        eval_metric="logloss",
        scale_pos_weight=(
            (y == 0).sum() /
            (y == 1).sum()
        )
    )
}

print("\nCross-validation configuration loaded successfully.")
````

````python
# =============================================================================
# STEP 22: 5-FOLD CV PREDICTIVE PERFORMANCE
# LASSO vs LASSO + DODA
# =============================================================================
# Feature selection is fitted inside each training fold.
# Scaling is also fitted inside each training fold.
#
# This prevents information from the validation fold from influencing
# LASSO feature selection, DODA ranking, or downstream Logistic Regression.

cv = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=RANDOM_STATE
)

cv_results = []


for method in [
    "LASSO",
    "LASSO + DODA"
]:

    print("\n" + "=" * 80)
    print(f"METHOD: {method}")
    print("=" * 80)

    for TOP_K in TOP_K_VALUES:

        for fold, (train_idx, val_idx) in enumerate(
            cv.split(X, y),
            start=1
        ):

            X_train_cv = X.iloc[train_idx].copy()
            X_val_cv = X.iloc[val_idx].copy()

            y_train_cv = y.iloc[train_idx].copy()
            y_val_cv = y.iloc[val_idx].copy()


            # -------------------------------------------------------------
            # SCALE WITH TRAINING-FOLD STATISTICS ONLY
            # -------------------------------------------------------------

            fold_scaler = StandardScaler()

            X_train_cv_scaled = pd.DataFrame(
                fold_scaler.fit_transform(X_train_cv),
                columns=X_train_cv.columns,
                index=X_train_cv.index
            )

            X_val_cv_scaled = pd.DataFrame(
                fold_scaler.transform(X_val_cv),
                columns=X_val_cv.columns,
                index=X_val_cv.index
            )


            # -------------------------------------------------------------
            # FEATURE SELECTION
            # -------------------------------------------------------------

            if method == "LASSO":

                lasso_selector_cv = make_lasso()

                lasso_selector_cv.fit(
                    X_train_cv_scaled,
                    y_train_cv
                )

                lasso_scores_cv = pd.DataFrame({
                    "Feature": X_train_cv.columns,
                    "LASSO_Coefficient":
                        lasso_selector_cv.coef_[0]
                })

                lasso_scores_cv["LASSO_Score"] = (
                    lasso_scores_cv[
                        "LASSO_Coefficient"
                    ].abs()
                )

                # Stable tie handling for reproducibility.
                lasso_scores_cv = (
                    lasso_scores_cv
                    .sort_values(
                        by=[
                            "LASSO_Score",
                            "Feature"
                        ],
                        ascending=[
                            False,
                            True
                        ]
                    )
                    .reset_index(drop=True)
                )

                selected_features = (
                    lasso_scores_cv
                    .head(TOP_K)["Feature"]
                    .tolist()
                )

            else:

                # DODA receives the complete scaled training matrix.
                # The LASSO operator supplies the statistical ranking,
                # while DODA applies the existing Hadamard fusion with
                # the clinical weights before selecting Top-K.

                doda_selector_cv = make_doda_lasso(
                    TOP_K
                )

                doda_selector_cv.fit(
                    X_train_cv_scaled,
                    y_train_cv
                )

                selected_features = list(
                    doda_selector_cv.get_selected_features()
                )


            # -------------------------------------------------------------
            # Selected train/validation matrices
            # -------------------------------------------------------------

            X_train_selected = X_train_cv_scaled[
                selected_features
            ]

            X_val_selected = X_val_cv_scaled[
                selected_features
            ]


            # -------------------------------------------------------------
            # Evaluate downstream models
            # -------------------------------------------------------------

            for model_name, model_template in cv_models.items():

                model = type(model_template)(
                    **model_template.get_params()
                )

                model.fit(
                    X_train_selected,
                    y_train_cv
                )

                y_pred = model.predict(
                    X_val_selected
                )

                y_prob = model.predict_proba(
                    X_val_selected
                )[:, 1]


                cv_results.append({

                    "Fold": fold,
                    "Method": method,
                    "Top_K": TOP_K,
                    "Model": model_name,

                    "Accuracy": accuracy_score(
                        y_val_cv,
                        y_pred
                    ),

                    "Precision": precision_score(
                        y_val_cv,
                        y_pred,
                        zero_division=0
                    ),

                    "Recall": recall_score(
                        y_val_cv,
                        y_pred,
                        zero_division=0
                    ),

                    "F1": f1_score(
                        y_val_cv,
                        y_pred,
                        zero_division=0
                    ),

                    "ROC_AUC": roc_auc_score(
                        y_val_cv,
                        y_prob
                    ),

                    "Selected_Features": ", ".join(
                        selected_features
                    )
                })


cv_results_df = pd.DataFrame(
    cv_results
)

print("\nCompleted 5-fold CV.")
print("Rows:", len(cv_results_df))

display(
    cv_results_df.head()
)
````

````python
# =============================================================================
# STEP 23: 5-FOLD CV SUMMARY — MEAN ± STD
# =============================================================================

cv_summary_df = (
    cv_results_df
    .groupby(
        ["Method", "Top_K", "Model"]
    )
    .agg(
        Accuracy_Mean=("Accuracy", "mean"),
        Accuracy_STD=("Accuracy", "std"),

        Precision_Mean=("Precision", "mean"),
        Precision_STD=("Precision", "std"),

        Recall_Mean=("Recall", "mean"),
        Recall_STD=("Recall", "std"),

        F1_Mean=("F1", "mean"),
        F1_STD=("F1", "std"),

        ROC_AUC_Mean=("ROC_AUC", "mean"),
        ROC_AUC_STD=("ROC_AUC", "std")
    )
    .reset_index()
)

print("=" * 100)
print("5-FOLD CV PERFORMANCE SUMMARY")
print("=" * 100)

display(
    cv_summary_df.round(4)
)

os.makedirs(
    "../../../results/breast_cancer/cv",
    exist_ok=True
)

cv_results_df.to_csv(
    "../../../results/breast_cancer/cv/lasso_doda_5fold_results.csv",
    index=False
)

cv_summary_df.to_csv(
    "../../../results/breast_cancer/cv/lasso_doda_5fold_summary.csv",
    index=False
)
````

````python
# =============================================================================
# STEP 24: 5×5 REPEATED STRATIFIED CV
# LASSO vs LASSO + DODA
# =============================================================================
# 5 folds × 5 repeats = 25 train/validation partitions.
# LASSO and DODA are refit inside every repeated-CV partition.
#
# Scaling is also refit inside every partition using training data only.
#
# The repeated-CV results are stored separately from the fixed-split
# baseline and DODA result tables.

repeated_cv = RepeatedStratifiedKFold(
    n_splits=N_SPLITS,
    n_repeats=N_REPEATS,
    random_state=RANDOM_STATE
)

repeated_cv_results = []


for run_id, (train_idx, val_idx) in enumerate(
    repeated_cv.split(X, y),
    start=1
):

    print("=" * 70)
    print(
        f"RUN {run_id}/{N_SPLITS * N_REPEATS}"
    )
    print("=" * 70)

    X_train_rep = X.iloc[train_idx].copy()
    X_val_rep = X.iloc[val_idx].copy()

    y_train_rep = y.iloc[train_idx].copy()
    y_val_rep = y.iloc[val_idx].copy()


    # ---------------------------------------------------------------------
    # Fold-specific scaling
    # ---------------------------------------------------------------------

    rep_scaler = StandardScaler()

    X_train_rep_scaled = pd.DataFrame(
        rep_scaler.fit_transform(X_train_rep),
        columns=X_train_rep.columns,
        index=X_train_rep.index
    )

    X_val_rep_scaled = pd.DataFrame(
        rep_scaler.transform(X_val_rep),
        columns=X_val_rep.columns,
        index=X_val_rep.index
    )


    # ---------------------------------------------------------------------
    # Fit LASSO once per repeated-CV partition.
    # The same LASSO ranking is reused across Top-K values.
    # ---------------------------------------------------------------------

    lasso_selector_rep = make_lasso()

    lasso_selector_rep.fit(
        X_train_rep_scaled,
        y_train_rep
    )

    lasso_rank_rep = pd.DataFrame({
        "Feature": X_train_rep.columns,
        "LASSO_Coefficient":
            lasso_selector_rep.coef_[0]
    })

    lasso_rank_rep["LASSO_Score"] = (
        lasso_rank_rep[
            "LASSO_Coefficient"
        ].abs()
    )

    lasso_rank_rep = (
        lasso_rank_rep
        .sort_values(
            by=[
                "LASSO_Score",
                "Feature"
            ],
            ascending=[
                False,
                True
            ]
        )
        .reset_index(drop=True)
    )


    for TOP_K in TOP_K_VALUES:

        # ---------------------------------------------------------------
        # LASSO baseline
        # ---------------------------------------------------------------

        lasso_features_rep = (
            lasso_rank_rep
            .head(TOP_K)["Feature"]
            .tolist()
        )


        # ---------------------------------------------------------------
        # LASSO + DODA
        # ---------------------------------------------------------------

        doda_selector_rep = make_doda_lasso(
            TOP_K
        )

        doda_selector_rep.fit(
            X_train_rep_scaled,
            y_train_rep
        )

        doda_features_rep = list(
            doda_selector_rep.get_selected_features()
        )


        feature_sets = {
            "LASSO": lasso_features_rep,
            "LASSO + DODA": doda_features_rep
        }


        for method, selected_features in feature_sets.items():

            for model_name, model_template in cv_models.items():

                model = type(model_template)(
                    **model_template.get_params()
                )

                Xtr = X_train_rep_scaled[
                    selected_features
                ]

                Xva = X_val_rep_scaled[
                    selected_features
                ]


                model.fit(
                    Xtr,
                    y_train_rep
                )

                y_pred = model.predict(
                    Xva
                )

                y_prob = model.predict_proba(
                    Xva
                )[:, 1]


                repeated_cv_results.append({

                    "Run": run_id,
                    "Method": method,
                    "Top_K": TOP_K,
                    "Model": model_name,

                    "Accuracy": accuracy_score(
                        y_val_rep,
                        y_pred
                    ),

                    "Precision": precision_score(
                        y_val_rep,
                        y_pred,
                        zero_division=0
                    ),

                    "Recall": recall_score(
                        y_val_rep,
                        y_pred,
                        zero_division=0
                    ),

                    "F1": f1_score(
                        y_val_rep,
                        y_pred,
                        zero_division=0
                    ),

                    "ROC_AUC": roc_auc_score(
                        y_val_rep,
                        y_prob
                    ),

                    "Selected_Features": ", ".join(
                        selected_features
                    )
                })


repeated_cv_results_df = pd.DataFrame(
    repeated_cv_results
)

print("\nCompleted repeated CV.")
print(
    "Rows:",
    len(repeated_cv_results_df)
)

display(
    repeated_cv_results_df.head()
)
````

````python
# =============================================================================
# STEP 25: REPEATED CV PERFORMANCE SUMMARY — MEAN ± STD
# =============================================================================

repeated_cv_summary_df = (
    repeated_cv_results_df
    .groupby(
        ["Method", "Top_K", "Model"]
    )
    .agg(
        Accuracy_Mean=("Accuracy", "mean"),
        Accuracy_STD=("Accuracy", "std"),

        Precision_Mean=("Precision", "mean"),
        Precision_STD=("Precision", "std"),

        Recall_Mean=("Recall", "mean"),
        Recall_STD=("Recall", "std"),

        F1_Mean=("F1", "mean"),
        F1_STD=("F1", "std"),

        ROC_AUC_Mean=("ROC_AUC", "mean"),
        ROC_AUC_STD=("ROC_AUC", "std")
    )
    .reset_index()
)

print("=" * 100)
print(
    "5×5 REPEATED STRATIFIED CV PERFORMANCE SUMMARY"
)
print("=" * 100)

display(
    repeated_cv_summary_df.round(4)
)

os.makedirs(
    "../../../results/breast_cancer/cv",
    exist_ok=True
)

repeated_cv_results_df.to_csv(
    "../../../results/breast_cancer/cv/"
    "lasso_doda_5x5_repeated_results.csv",
    index=False
)

repeated_cv_summary_df.to_csv(
    "../../../results/breast_cancer/cv/"
    "lasso_doda_5x5_repeated_summary.csv",
    index=False
)
````

## 4. Feature-Selection Stability and LASSO–DODA Agreement

The next analyses use the **25 repeated-CV runs** to examine how consistently features are selected.

We distinguish:

- **Selection frequency:** how often a feature is selected.
- **Jaccard similarity:** overlap between feature sets.
- **Kuncheva index:** chance-adjusted agreement for fixed-size feature subsets.
- **LASSO–DODA agreement:** whether DODA changes the selected set relative to the LASSO statistical baseline.

These analyses evaluate **feature-selection stability and set membership**, not merely predictive performance.


````python
# =============================================================================
# STEP 26: PREPARE STABILITY DATAFRAME
# =============================================================================

stability_df = repeated_cv_results_df[
    [
        "Run",
        "Method",
        "Top_K",
        "Selected_Features"
    ]
].copy()

stability_df["Feature_Set"] = (
    stability_df["Selected_Features"]
    .apply(
        lambda x: {
            feature.strip()
            for feature in x.split(",")
            if feature.strip()
        }
    )
)

feature_names = X.columns.tolist()
N_FEATURES = len(feature_names)

print("=" * 80)
print("STABILITY ANALYSIS PREPARATION")
print("=" * 80)

print("Rows:", stability_df.shape[0])
print(
    "Methods:",
    stability_df["Method"].unique()
)
print(
    "Top-K:",
    sorted(stability_df["Top_K"].unique())
)
print(
    "Total predictors:",
    N_FEATURES
)

display(
    stability_df[
        [
            "Run",
            "Method",
            "Top_K",
            "Feature_Set"
        ]
    ].head()
)
````

````python
# =============================================================================
# STEP 27: TEST 1 — FEATURE SELECTION FREQUENCY
# =============================================================================

selection_frequency = []


for method in sorted(
    stability_df["Method"].unique()
):

    for top_k in sorted(
        stability_df["Top_K"].unique()
    ):

        subset = stability_df[
            (stability_df["Method"] == method) &
            (stability_df["Top_K"] == top_k)
        ]

        total_runs = len(subset)

        all_features = sorted(
            set().union(
                *subset["Feature_Set"]
            )
        )

        for feature in all_features:

            selected_count = sum(
                feature in feature_set
                for feature_set in subset["Feature_Set"]
            )

            selection_frequency.append({

                "Method": method,

                "Top_K": top_k,

                "Feature": feature,

                "Selected_Count": selected_count,

                "Total_Runs": total_runs,

                "Selection_Frequency_Percentage":
                    (
                        selected_count /
                        total_runs
                    ) * 100
            })


selection_frequency_df = pd.DataFrame(
    selection_frequency
)

selection_frequency_df = (
    selection_frequency_df
    .sort_values(
        by=[
            "Method",
            "Top_K",
            "Selection_Frequency_Percentage"
        ],
        ascending=[
            True,
            True,
            False
        ]
    )
    .reset_index(drop=True)
)

display(
    selection_frequency_df
)
````

````python
# =============================================================================
# STEP 28: TOP FREQUENTLY SELECTED FEATURES
# =============================================================================

top_frequency_features = (
    selection_frequency_df
    .groupby(
        ["Method", "Top_K"],
        group_keys=False
    )
    .head(10)
    .reset_index(drop=True)
)

print("=" * 80)
print("TOP FREQUENTLY SELECTED FEATURES")
print("=" * 80)

display(
    top_frequency_features
)
````

````python
# =============================================================================
# STEP 29: TEST 2 — PAIRWISE JACCARD SIMILARITY
# =============================================================================

jaccard_results = []


for method in sorted(
    stability_df["Method"].unique()
):

    for top_k in sorted(
        stability_df["Top_K"].unique()
    ):

        subset = stability_df[
            (stability_df["Method"] == method) &
            (stability_df["Top_K"] == top_k)
        ].reset_index(drop=True)

        feature_sets = (
            subset["Feature_Set"]
            .tolist()
        )

        pairwise_scores = []


        for set_a, set_b in combinations(
            feature_sets,
            2
        ):

            intersection_size = len(
                set_a.intersection(set_b)
            )

            union_size = len(
                set_a.union(set_b)
            )

            jaccard_score = (
                intersection_size /
                union_size
                if union_size
                else 1.0
            )

            pairwise_scores.append(
                jaccard_score
            )


        if pairwise_scores:

            jaccard_results.append({

                "Method": method,

                "Top_K": top_k,

                "Mean_Jaccard": np.mean(
                    pairwise_scores
                ),

                "Std_Jaccard": np.std(
                    pairwise_scores,
                    ddof=1
                ),

                "Minimum_Jaccard": np.min(
                    pairwise_scores
                ),

                "Maximum_Jaccard": np.max(
                    pairwise_scores
                )
            })


jaccard_df = pd.DataFrame(
    jaccard_results
)

print("=" * 80)
print("PAIRWISE JACCARD SIMILARITY")
print("=" * 80)

display(
    jaccard_df.round(4)
)
````

````python
# =============================================================================
# STEP 30: TEST 3 — KUNCHEVA INDEX
# =============================================================================

kuncheva_results = []


for method in sorted(
    stability_df["Method"].unique()
):

    for top_k in sorted(
        stability_df["Top_K"].unique()
    ):

        subset = stability_df[
            (stability_df["Method"] == method) &
            (stability_df["Top_K"] == top_k)
        ].reset_index(drop=True)

        feature_sets = (
            subset["Feature_Set"]
            .tolist()
        )


        for i in range(
            len(feature_sets)
        ):

            for j in range(
                i + 1,
                len(feature_sets)
            ):

                set_a = feature_sets[i]
                set_b = feature_sets[j]

                intersection = len(
                    set_a.intersection(set_b)
                )

                denominator = (
                    top_k *
                    (
                        N_FEATURES -
                        top_k
                    )
                )


                if denominator == 0:

                    kuncheva_value = np.nan

                else:

                    kuncheva_value = (
                        (
                            intersection *
                            N_FEATURES
                        )
                        -
                        (
                            top_k ** 2
                        )
                    ) / denominator


                kuncheva_results.append({

                    "Method": method,

                    "Top_K": top_k,

                    "Run_A": subset.loc[
                        i,
                        "Run"
                    ],

                    "Run_B": subset.loc[
                        j,
                        "Run"
                    ],

                    "Intersection": intersection,

                    "Kuncheva_Index": kuncheva_value
                })


kuncheva_df = pd.DataFrame(
    kuncheva_results
)

kuncheva_summary = (
    kuncheva_df
    .groupby(
        ["Method", "Top_K"]
    )["Kuncheva_Index"]
    .agg(
        Mean_Kuncheva="mean",
        Std_Kuncheva="std",
        Min_Kuncheva="min",
        Max_Kuncheva="max"
    )
    .reset_index()
)


print("=" * 80)
print("KUNCHEVA INDEX SUMMARY")
print("=" * 80)

display(
    kuncheva_summary.round(4)
)
````

````python
# =============================================================================
# STEP 31: TEST 4 — LASSO vs DODA FEATURE-SET AGREEMENT
# =============================================================================
# This compares feature-set membership within the SAME repeated-CV run
# and Top-K value.

agreement_results = []


for run in sorted(
    stability_df["Run"].unique()
):

    for top_k in sorted(
        stability_df["Top_K"].unique()
    ):

        lasso_rows = stability_df[
            (stability_df["Run"] == run) &
            (stability_df["Method"] == "LASSO") &
            (stability_df["Top_K"] == top_k)
        ]

        doda_rows = stability_df[
            (stability_df["Run"] == run) &
            (stability_df["Method"] == "LASSO + DODA") &
            (stability_df["Top_K"] == top_k)
        ]


        if (
            lasso_rows.empty or
            doda_rows.empty
        ):
            continue


        lasso_set = (
            lasso_rows.iloc[0]["Feature_Set"]
        )

        doda_set = (
            doda_rows.iloc[0]["Feature_Set"]
        )


        intersection_size = len(
            lasso_set.intersection(
                doda_set
            )
        )

        union_size = len(
            lasso_set.union(
                doda_set
            )
        )

        jaccard_agreement = (
            intersection_size /
            union_size
            if union_size
            else 1.0
        )


        agreement_results.append({

            "Run": run,

            "Top_K": top_k,

            "LASSO_Features":
                ", ".join(
                    sorted(lasso_set)
                ),

            "DODA_Features":
                ", ".join(
                    sorted(doda_set)
                ),

            "Common_Feature_Count":
                intersection_size,

            "LASSO_Feature_Count":
                len(lasso_set),

            "DODA_Feature_Count":
                len(doda_set),

            "Same_Feature_Set":
                lasso_set == doda_set,

            "Jaccard_Agreement":
                jaccard_agreement,

            "Features_Added_by_DODA":
                ", ".join(
                    sorted(
                        doda_set -
                        lasso_set
                    )
                ),

            "Features_Removed_by_DODA":
                ", ".join(
                    sorted(
                        lasso_set -
                        doda_set
                    )
                )
        })


agreement_df = pd.DataFrame(
    agreement_results
)


print("=" * 100)
print("LASSO vs DODA FEATURE-SET AGREEMENT")
print("=" * 100)

display(
    agreement_df
)
````

````python
# =============================================================================
# STEP 32: SUMMARY OF LASSO vs DODA AGREEMENT
# =============================================================================

agreement_summary_df = (
    agreement_df
    .groupby("Top_K")
    .agg(
        Same_Feature_Set_Percentage=(
            "Same_Feature_Set",
            lambda x: x.mean() * 100
        ),

        Mean_Jaccard_Agreement=(
            "Jaccard_Agreement",
            "mean"
        ),

        Std_Jaccard_Agreement=(
            "Jaccard_Agreement",
            "std"
        ),

        Mean_Common_Features=(
            "Common_Feature_Count",
            "mean"
        )
    )
    .reset_index()
)


print("=" * 100)
print(
    "FINAL LASSO vs DODA FEATURE-SET AGREEMENT SUMMARY"
)
print("=" * 100)

display(
    agreement_summary_df.round(4)
)


# -------------------------------------------------------------------------
# Save stability outputs
# -------------------------------------------------------------------------

os.makedirs(
    "../../../results/breast_cancer/stability",
    exist_ok=True
)

selection_frequency_df.to_csv(
    "../../../results/breast_cancer/stability/"
    "feature_selection_frequency_lasso.csv",
    index=False
)

jaccard_df.to_csv(
    "../../../results/breast_cancer/stability/"
    "jaccard_similarity_lasso.csv",
    index=False
)

kuncheva_df.to_csv(
    "../../../results/breast_cancer/stability/"
    "kuncheva_pairwise_lasso.csv",
    index=False
)

kuncheva_summary.to_csv(
    "../../../results/breast_cancer/stability/"
    "kuncheva_summary_lasso.csv",
    index=False
)

agreement_df.to_csv(
    "../../../results/breast_cancer/stability/"
    "lasso_doda_agreement.csv",
    index=False
)

agreement_summary_df.to_csv(
    "../../../results/breast_cancer/stability/"
    "lasso_doda_agreement_summary.csv",
    index=False
)
````
