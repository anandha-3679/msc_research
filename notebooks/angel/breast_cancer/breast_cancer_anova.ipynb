## 1. Baseline

```python
# =============================================================================
# ANOVA BASELINE FEATURE SELECTION (TOP-K = 5,10,15,20)
# =============================================================================

from sklearn.feature_selection import SelectKBest, f_classif
import pandas as pd


# Fit ANOVA once with all features
anova_selector = SelectKBest(
    score_func=f_classif,

    k="all"
)

anova_selector.fit(
    X_train_scaled,
    y_train
)


# Create score dataframe
anova_scores = pd.DataFrame({

    "Feature": X_train_scaled.columns,

    "ANOVA Score": anova_selector.scores_

})


# Sort by ANOVA importance
anova_scores = anova_scores.sort_values(
    by="ANOVA Score",
    ascending=False
).reset_index(drop=True)



# Top-k values
k_values = [5, 10, 15, 20]

anova_results = {}


for k in k_values:

    top_features = (
        anova_scores
        .head(k)["Feature"]
        .tolist()
    )


    # Boolean mask for selecting columns
    mask = X_train_scaled.columns.isin(
        top_features
    )


    X_train_selected = X_train_scaled.loc[
        :, mask
    ]

    X_test_selected = X_test_scaled.loc[
        :, mask
    ]


    anova_results[k] = {

        "features": top_features,

        "X_train": X_train_selected,

        "X_test": X_test_selected,

        "scores": anova_scores.head(k)

    }


    print("="*70)
    print(f"TOP {k} ANOVA FEATURES")
    print("="*70)

    print(top_features)

    print("\nScores:")
    print(anova_scores.head(k))
```

```python
# =============================================================================
# BASELINE MODELS
# =============================================================================

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier


models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        random_state=42,

        eval_metric="logloss",
        scale_pos_weight=(y_train == 0).sum() / (y_train == 1).sum()
    )
}
```

```python
# =============================================================================
# MODEL EVALUATION
# =============================================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

results = []

for k in k_values:

    Xtr = anova_results[k]["X_train"]
    Xts = anova_results[k]["X_test"]

    features = ", ".join(anova_results[k]["features"])

    for model_name, model in models.items():

        model.fit(Xtr, y_train)

        y_pred = model.predict(Xts)

        y_prob = model.predict_proba(Xts)[:,1]

        results.append({

            "Method": "ANOVA",

            "Top-K": k,

            "Model": model_name,

            "Selected Features": features,

            "Accuracy": accuracy_score(y_test, y_pred),

            "Precision": precision_score(y_test, y_pred),

            "Recall": recall_score(y_test, y_pred),

            "F1 Score": f1_score(y_test, y_pred),

            "ROC-AUC": roc_auc_score(y_test, y_prob)

        })
```

```python
results_df = pd.DataFrame(results)

results_df
```

```python
# =============================================================================
# ANOVA + DODA FEATURE SELECTION
# =============================================================================

import pandas as pd
from sklearn.feature_selection import SelectKBest, f_classif

# -------------------------------------------------------------------------
# Top-K values
# -------------------------------------------------------------------------

k_values = [5, 10, 15, 20]


# -------------------------------------------------------------------------
# Clinical knowledge provider
# -------------------------------------------------------------------------

provider = JSONProvider(
    "../../../config/clinical_weights/breast_cancer.json"
)


# -------------------------------------------------------------------------
# Fusion method
# -------------------------------------------------------------------------

fusion = HadamardFusion()


# -------------------------------------------------------------------------
# Store DODA results
# -------------------------------------------------------------------------

doda_results = {}


# =============================================================================
# RUN DODA FOR EACH TOP-K
# =============================================================================

for k in k_values:

    print("\n" + "=" * 70)
    print(f"ANOVA + DODA \u2014 TOP {k} FEATURES")
    print("=" * 70)


    # ---------------------------------------------------------------------
    # ANOVA operator
    # ---------------------------------------------------------------------

    anova = SklearnAdapter(
        SelectKBest(
            score_func=f_classif,
            k=k
        )
    )


    # ---------------------------------------------------------------------
    # DODA selector
    # ---------------------------------------------------------------------

    selector = DODASelector(
        operators=[anova],
        provider=provider,
        fusion=fusion,
        top_k=k
    )


    # ---------------------------------------------------------------------
    # Fit DODA on TRAINING data
    # ---------------------------------------------------------------------

    X_train_selected = selector.fit_transform(
        X_train_scaled,
        y_train
    )


    # ---------------------------------------------------------------------
    # Transform TEST data
    # ---------------------------------------------------------------------

    X_test_selected = selector.transform(
        X_test_scaled
    )


    # ---------------------------------------------------------------------
    # Selected features
    # ---------------------------------------------------------------------

    selected_features = selector.get_selected_features()


    print("\nSelected Features:")
    print(selected_features)


    # ---------------------------------------------------------------------
    # Store everything needed for later evaluation
    # ---------------------------------------------------------------------

    doda_results[k] = {

        "selector": selector,

        "X_train": X_train_selected,

        "X_test": X_test_selected,

        "features": selected_features,

        "raw_math_scores": selector.raw_math_scores_,

        "math_scores": selector.math_scores_,

        "clinical_weights": selector.clinical_weights_,

        "final_scores": selector.final_scores_
    }


print("\n" + "=" * 70)
print("DODA FEATURE SELECTION COMPLETED")
print("=" * 70)
```

```python
# CELL 2
# =============================================================================
# INSPECT MATHEMATICAL SCORES
# =============================================================================

k = 10

selector = doda_results[k]["selector"]

# -----------------------------------------------------------------------------
# Raw mathematical scores
# -----------------------------------------------------------------------------

raw_math_scores = pd.DataFrame(
    selector.raw_math_scores_.items(),
    columns=[
        "Feature",
        "Raw Math Score"
    ]
)

# -----------------------------------------------------------------------------
# Normalized mathematical scores
# -----------------------------------------------------------------------------

normalized_math_scores = pd.DataFrame(
    selector.math_scores_.items(),
    columns=[
        "Feature",
        "Normalized Math Score"
    ]
)

# -----------------------------------------------------------------------------
# Combine both
# -----------------------------------------------------------------------------

math_scores = raw_math_scores.merge(
    normalized_math_scores,
    on="Feature"
)

math_scores = math_scores.sort_values(
    by="Normalized Math Score",
    ascending=False
)

print("=" * 70)
print("MATHEMATICAL SCORES")
print("=" * 70)

display(math_scores)
```

```python
# CELL 3
# =============================================================================
# INSPECT CLINICAL WEIGHTS
# =============================================================================

clinical_weights = pd.DataFrame(
    selector.clinical_weights_.items(),
    columns=[
        "Feature",
        "Clinical Weight"
    ]
)

clinical_weights = clinical_weights.sort_values(
    by="Clinical Weight",
    ascending=False
)

print("=" * 70)
print("CLINICAL WEIGHTS")
print("=" * 70)

display(clinical_weights)
```

```python
# CELL 4
# =============================================================================
# INSPECT FINAL DODA SCORES
# =============================================================================

final_scores = pd.DataFrame(
    selector.final_scores_.items(),
    columns=[
        "Feature",
        "Final DODA Score"
    ]
)

final_scores = final_scores.sort_values(
    by="Final DODA Score",
    ascending=False
)

print("=" * 70)
print("FINAL DODA SCORES")
print("=" * 70)

display(final_scores)
```

## 4. Cross-Validation and Feature-Selection Stability

The single 80/20 train-test experiment above is retained as the fixed-split analysis.  
The following sections add cross-validation and stability analyses to strengthen the breast-cancer experiment.

The purpose is to assess whether the observed ANOVA/DODA feature-selection and predictive results persist across different data partitions rather than relying only on `random_state=42`.

**Important:** feature selection is refit inside every training fold. Validation/test folds are not used to determine ANOVA scores or DODA-selected features.


```python
# =============================================================================
# STEP 21: CROSS-VALIDATION CONFIGURATION
# =============================================================================

import numpy as np
import pandas as pd

from itertools import combinations

from sklearn.model_selection import StratifiedKFold, RepeatedStratifiedKFold
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

from doda import DODASelector
from doda.adapters.sklearn_adapter import SklearnAdapter
from doda.knowledge.providers import JSONProvider
from doda.fusion import HadamardFusion


# -------------------------------------------------------------------------
# Research design
# -------------------------------------------------------------------------

TOP_K_VALUES = [5, 10, 15, 20]

N_SPLITS = 5
N_REPEATS = 5
RANDOM_STATE = 42

WEIGHTS_FILE = "../../../config/clinical_weights/breast_cancer.json"

print("=" * 80)
print("BREAST CANCER CROSS-VALIDATION CONFIGURATION")
print("=" * 80)

print("Top-K values:", TOP_K_VALUES)
print("5-fold CV:", N_SPLITS)
print("Repeated CV:", f"{N_SPLITS} folds × {N_REPEATS} repeats")
print("Total repeated runs:", N_SPLITS * N_REPEATS)
print("Random state:", RANDOM_STATE)
print("Clinical weights:", WEIGHTS_FILE)
```

```python
# =============================================================================
# STEP 22: 5-FOLD CV PREDICTIVE PERFORMANCE
# ANOVA vs ANOVA + DODA
# =============================================================================
# This section extends the breast-cancer experiment with cross-validation.
# Unlike the original template, it does NOT perform imputation because the
# breast-cancer processed dataset is checked for missing values above.
#
# Feature selection is fitted inside each training fold.

cv_models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=RANDOM_STATE
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=RANDOM_STATE
    ),

    "XGBoost": XGBClassifier(
        random_state=RANDOM_STATE,
        eval_metric="logloss",
        scale_pos_weight=(
            (y == 0).sum() / (y == 1).sum()
        )
    )
}

cv = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=RANDOM_STATE
)

cv_results = []

for method in ["ANOVA", "DODA"]:

    print("\n" + "=" * 80)
    print(f"METHOD: {method}")
    print("=" * 80)

    for TOP_K in TOP_K_VALUES:

        for model_name, base_model in cv_models.items():

            for fold, (train_idx, val_idx) in enumerate(
                cv.split(X, y),
                start=1
            ):

                X_train_cv = X.iloc[train_idx].copy()
                X_val_cv = X.iloc[val_idx].copy()

                y_train_cv = y.iloc[train_idx].copy()
                y_val_cv = y.iloc[val_idx].copy()

                # ---------------------------------------------------------
                # FEATURE SELECTION
                # ---------------------------------------------------------

                if method == "ANOVA":

                    selector_cv = SelectKBest(
                        score_func=f_classif,
                        k=TOP_K
                    )

                    selector_cv.fit(
                        X_train_cv,
                        y_train_cv
                    )

                    selected_features = (
                        X_train_cv.columns[
                            selector_cv.get_support()
                        ].tolist()
                    )

                else:

                    # ANOVA scores ALL candidate features.
                    # DODA performs final Top-K selection.
                    anova_operator = SklearnAdapter(
                        SelectKBest(
                            score_func=f_classif,
                            k="all"
                        )
                    )

                    provider_cv = JSONProvider(
                        WEIGHTS_FILE
                    )

                    doda_selector_cv = DODASelector(
                        operators=[anova_operator],
                        provider=provider_cv,
                        fusion=HadamardFusion(),
                        top_k=TOP_K
                    )

                    doda_selector_cv.fit(
                        X_train_cv,
                        y_train_cv
                    )

                    selected_features = list(
                        doda_selector_cv.get_selected_features()
                    )

                X_train_selected = X_train_cv[
                    selected_features
                ]

                X_val_selected = X_val_cv[
                    selected_features
                ]

                # ---------------------------------------------------------
                # FRESH MODEL FOR THIS FOLD
                # ---------------------------------------------------------

                model = type(base_model)(
                    **base_model.get_params()
                )

                Xtr = X_train_selected
                Xva = X_val_selected

                # Only Logistic Regression requires scaling.
                if model_name == "Logistic Regression":

                    fold_scaler = StandardScaler()

                    Xtr = fold_scaler.fit_transform(Xtr)
                    Xva = fold_scaler.transform(Xva)

                model.fit(
                    Xtr,
                    y_train_cv
                )

                y_pred = model.predict(Xva)
                y_prob = model.predict_proba(Xva)[:, 1]

                cv_results.append({

                    "Fold": fold,
                    "Method": method,
                    "Top_K": TOP_K,
                    "Model": model_name,

                    "Accuracy": accuracy_score(
                        y_val_cv,
                        y_pred
                    ),

                    "Precision": precision_score(
                        y_val_cv,
                        y_pred,
                        zero_division=0
                    ),

                    "Recall": recall_score(
                        y_val_cv,
                        y_pred,
                        zero_division=0
                    ),

                    "F1": f1_score(
                        y_val_cv,
                        y_pred,
                        zero_division=0
                    ),

                    "ROC_AUC": roc_auc_score(
                        y_val_cv,
                        y_prob
                    ),

                    "Selected_Features": ", ".join(
                        selected_features
                    )
                })

cv_results_df = pd.DataFrame(cv_results)

print("\nCompleted 5-fold CV.")
print("Rows:", len(cv_results_df))

display(cv_results_df.head())
```

```python
# =============================================================================
# STEP 23: 5-FOLD CV SUMMARY — MEAN ± STD
# =============================================================================

cv_summary_df = (
    cv_results_df
    .groupby(
        ["Method", "Top_K", "Model"]
    )
    .agg(
        Accuracy_Mean=("Accuracy", "mean"),
        Accuracy_STD=("Accuracy", "std"),

        Precision_Mean=("Precision", "mean"),
        Precision_STD=("Precision", "std"),

        Recall_Mean=("Recall", "mean"),
        Recall_STD=("Recall", "std"),

        F1_Mean=("F1", "mean"),
        F1_STD=("F1", "std"),

        ROC_AUC_Mean=("ROC_AUC", "mean"),
        ROC_AUC_STD=("ROC_AUC", "std")
    )
    .reset_index()
)

print("=" * 100)
print("5-FOLD CV PERFORMANCE SUMMARY")
print("=" * 100)

display(cv_summary_df.round(4))


# -------------------------------------------------------------------------
# Save
# -------------------------------------------------------------------------

os.makedirs(
    "../../../results/breast_cancer/cv",
    exist_ok=True
)

cv_results_df.to_csv(
    "../../../results/breast_cancer/cv/anova_doda_5fold_results.csv",
    index=False
)

cv_summary_df.to_csv(
    "../../../results/breast_cancer/cv/anova_doda_5fold_summary.csv",
    index=False
)
```

```python
# =============================================================================
# STEP 24: 5×5 REPEATED STRATIFIED CV
# ANOVA vs ANOVA + DODA
# =============================================================================
# 5 folds × 5 repeats = 25 independent train/validation partitions.
# Feature selection is refit within every run.
#
# No imputation is performed because the dataset was verified to contain
# no missing values.
#
# The repeated-CV results are kept in a separate dataframe so that they
# do not overwrite the fixed-split baseline results_df.

repeated_cv = RepeatedStratifiedKFold(
    n_splits=N_SPLITS,
    n_repeats=N_REPEATS,
    random_state=RANDOM_STATE
)

repeated_cv_results = []

for run_id, (train_idx, val_idx) in enumerate(
    repeated_cv.split(X, y),
    start=1
):

    print("=" * 70)
    print(f"RUN {run_id}/{N_SPLITS * N_REPEATS}")
    print("=" * 70)

    X_train_rep = X.iloc[train_idx].copy()
    X_val_rep = X.iloc[val_idx].copy()

    y_train_rep = y.iloc[train_idx].copy()
    y_val_rep = y.iloc[val_idx].copy()

    for TOP_K in TOP_K_VALUES:

        # ---------------------------------------------------------------
        # ANOVA baseline
        # ---------------------------------------------------------------

        anova_selector_rep = SelectKBest(
            score_func=f_classif,
            k=TOP_K
        )

        anova_selector_rep.fit(
            X_train_rep,
            y_train_rep
        )

        anova_features_rep = (
            X_train_rep.columns[
                anova_selector_rep.get_support()
            ].tolist()
        )

        # ---------------------------------------------------------------
        # DODA
        # ---------------------------------------------------------------

        anova_operator_rep = SklearnAdapter(
            SelectKBest(
                score_func=f_classif,
                k="all"
            )
        )

        provider_rep = JSONProvider(
            WEIGHTS_FILE
        )

        doda_selector_rep = DODASelector(
            operators=[anova_operator_rep],
            provider=provider_rep,
            fusion=HadamardFusion(),
            top_k=TOP_K
        )

        doda_selector_rep.fit(
            X_train_rep,
            y_train_rep
        )

        doda_features_rep = list(
            doda_selector_rep.get_selected_features()
        )

        # ---------------------------------------------------------------
        # Evaluate both feature-selection conditions with each model
        # ---------------------------------------------------------------

        for model_name, model_template in cv_models.items():

            # =========================
            # ANOVA
            # =========================

            model = type(model_template)(
                **model_template.get_params()
            )

            Xtr = X_train_rep[
                anova_features_rep
            ]

            Xva = X_val_rep[
                anova_features_rep
            ]

            if model_name == "Logistic Regression":

                rep_scaler = StandardScaler()

                Xtr = rep_scaler.fit_transform(Xtr)
                Xva = rep_scaler.transform(Xva)

            model.fit(
                Xtr,
                y_train_rep
            )

            y_pred = model.predict(Xva)
            y_prob = model.predict_proba(Xva)[:, 1]

            repeated_cv_results.append({

                "Run": run_id,
                "Method": "ANOVA",
                "Top_K": TOP_K,
                "Model": model_name,

                "Accuracy": accuracy_score(
                    y_val_rep,
                    y_pred
                ),

                "Precision": precision_score(
                    y_val_rep,
                    y_pred,
                    zero_division=0
                ),

                "Recall": recall_score(
                    y_val_rep,
                    y_pred,
                    zero_division=0
                ),

                "F1": f1_score(
                    y_val_rep,
                    y_pred,
                    zero_division=0
                ),

                "ROC_AUC": roc_auc_score(
                    y_val_rep,
                    y_prob
                ),

                "Selected_Features": ", ".join(
                    anova_features_rep
                )
            })

            # =========================
            # DODA
            # =========================

            model = type(model_template)(
                **model_template.get_params()
            )

            Xtr = X_train_rep[
                doda_features_rep
            ]

            Xva = X_val_rep[
                doda_features_rep
            ]

            if model_name == "Logistic Regression":

                rep_scaler = StandardScaler()

                Xtr = rep_scaler.fit_transform(Xtr)
                Xva = rep_scaler.transform(Xva)

            model.fit(
                Xtr,
                y_train_rep
            )

            y_pred = model.predict(Xva)
            y_prob = model.predict_proba(Xva)[:, 1]

            repeated_cv_results.append({

                "Run": run_id,
                "Method": "DODA",
                "Top_K": TOP_K,
                "Model": model_name,

                "Accuracy": accuracy_score(
                    y_val_rep,
                    y_pred
                ),

                "Precision": precision_score(
                    y_val_rep,
                    y_pred,
                    zero_division=0
                ),

                "Recall": recall_score(
                    y_val_rep,
                    y_pred,
                    zero_division=0
                ),

                "F1": f1_score(
                    y_val_rep,
                    y_pred,
                    zero_division=0
                ),

                "ROC_AUC": roc_auc_score(
                    y_val_rep,
                    y_prob
                ),

                "Selected_Features": ", ".join(
                    doda_features_rep
                )
            })

repeated_cv_results_df = pd.DataFrame(
    repeated_cv_results
)

print("\nCompleted repeated CV.")
print("Rows:", len(repeated_cv_results_df))

display(
    repeated_cv_results_df.head()
)
```

```python
# =============================================================================
# STEP 25: REPEATED CV PERFORMANCE SUMMARY — MEAN ± STD
# =============================================================================

repeated_cv_summary_df = (
    repeated_cv_results_df
    .groupby(
        ["Method", "Top_K", "Model"]
    )
    .agg(
        Accuracy_Mean=("Accuracy", "mean"),
        Accuracy_STD=("Accuracy", "std"),

        Precision_Mean=("Precision", "mean"),
        Precision_STD=("Precision", "std"),

        Recall_Mean=("Recall", "mean"),
        Recall_STD=("Recall", "std"),

        F1_Mean=("F1", "mean"),
        F1_STD=("F1", "std"),

        ROC_AUC_Mean=("ROC_AUC", "mean"),
        ROC_AUC_STD=("ROC_AUC", "std")
    )
    .reset_index()
)

print("=" * 100)
print("5×5 REPEATED STRATIFIED CV PERFORMANCE SUMMARY")
print("=" * 100)

display(
    repeated_cv_summary_df.round(4)
)

repeated_cv_results_df.to_csv(
    "../../../results/breast_cancer/cv/anova_doda_5x5_repeated_results.csv",
    index=False
)

repeated_cv_summary_df.to_csv(
    "../../../results/breast_cancer/cv/anova_doda_5x5_repeated_summary.csv",
    index=False
)
```

## 5. Feature-Selection Stability and ANOVA–DODA Agreement

The next analyses use the **25 repeated-CV runs** to examine how consistently features are selected.

We distinguish:

- **Selection frequency:** how often a feature is selected.
- **Jaccard similarity:** overlap between feature sets.
- **Kuncheva index:** chance-adjusted agreement for fixed-size feature subsets.
- **ANOVA–DODA agreement:** whether DODA changes the selected set relative to the statistical baseline.


```python
# =============================================================================
# STEP 26: PREPARE STABILITY DATAFRAME
# =============================================================================

stability_df = repeated_cv_results_df[
    [
        "Run",
        "Method",
        "Top_K",
        "Selected_Features"
    ]
].copy()

stability_df["Feature_Set"] = (
    stability_df["Selected_Features"]
    .apply(
        lambda x: {
            feature.strip()
            for feature in x.split(",")
            if feature.strip()
        }
    )
)

feature_names = X.columns.tolist()
N_FEATURES = len(feature_names)

print("=" * 80)
print("STABILITY ANALYSIS PREPARATION")
print("=" * 80)

print("Rows:", stability_df.shape[0])
print("Methods:", stability_df["Method"].unique())
print("Top-K:", sorted(stability_df["Top_K"].unique()))
print("Total predictors:", N_FEATURES)

display(
    stability_df[
        [
            "Run",
            "Method",
            "Top_K",
            "Feature_Set"
        ]
    ].head()
)
```

```python
# =============================================================================
# STEP 27: TEST 1 — FEATURE SELECTION FREQUENCY
# =============================================================================

selection_frequency = []

for method in sorted(
    stability_df["Method"].unique()
):

    for top_k in sorted(
        stability_df["Top_K"].unique()
    ):

        subset = stability_df[
            (stability_df["Method"] == method) &
            (stability_df["Top_K"] == top_k)
        ]

        total_runs = len(subset)

        all_features = sorted(
            set().union(
                *subset["Feature_Set"]
            )
        )

        for feature in all_features:

            selected_count = sum(
                feature in feature_set
                for feature_set in subset["Feature_Set"]
            )

            selection_frequency.append({

                "Method": method,

                "Top_K": top_k,

                "Feature": feature,

                "Selected_Count": selected_count,

                "Total_Runs": total_runs,

                "Selection_Frequency_Percentage":
                    (
                        selected_count /
                        total_runs
                    ) * 100
            })

selection_frequency_df = pd.DataFrame(
    selection_frequency
)

selection_frequency_df = (
    selection_frequency_df
    .sort_values(
        by=[
            "Method",
            "Top_K",
            "Selection_Frequency_Percentage"
        ],
        ascending=[
            True,
            True,
            False
        ]
    )
    .reset_index(drop=True)
)

display(selection_frequency_df)
```

```python
# =============================================================================
# STEP 28: TOP FREQUENTLY SELECTED FEATURES
# =============================================================================

top_frequency_features = (
    selection_frequency_df
    .groupby(
        ["Method", "Top_K"],
        group_keys=False
    )
    .head(10)
    .reset_index(drop=True)
)

print("=" * 80)
print("TOP FREQUENTLY SELECTED FEATURES")
print("=" * 80)

display(top_frequency_features)
```

```python
# =============================================================================
# STEP 29: TEST 2 — PAIRWISE JACCARD SIMILARITY
# =============================================================================

jaccard_results = []

for method in sorted(
    stability_df["Method"].unique()
):

    for top_k in sorted(
        stability_df["Top_K"].unique()
    ):

        subset = stability_df[
            (stability_df["Method"] == method) &
            (stability_df["Top_K"] == top_k)
        ].reset_index(drop=True)

        feature_sets = (
            subset["Feature_Set"]
            .tolist()
        )

        pairwise_scores = []

        for set_a, set_b in combinations(
            feature_sets,
            2
        ):

            intersection_size = len(
                set_a.intersection(set_b)
            )

            union_size = len(
                set_a.union(set_b)
            )

            jaccard_score = (
                intersection_size / union_size
                if union_size
                else 1.0
            )

            pairwise_scores.append(
                jaccard_score
            )

        if pairwise_scores:

            jaccard_results.append({

                "Method": method,

                "Top_K": top_k,

                "Mean_Jaccard": np.mean(
                    pairwise_scores
                ),

                "Std_Jaccard": np.std(
                    pairwise_scores,
                    ddof=1
                ),

                "Minimum_Jaccard": np.min(
                    pairwise_scores
                ),

                "Maximum_Jaccard": np.max(
                    pairwise_scores
                )
            })

jaccard_df = pd.DataFrame(
    jaccard_results
)

print("=" * 80)
print("PAIRWISE JACCARD SIMILARITY")
print("=" * 80)

display(jaccard_df.round(4))
```

```python
# =============================================================================
# STEP 30: TEST 3 — KUNCHEVA INDEX
# =============================================================================

kuncheva_results = []

for method in sorted(
    stability_df["Method"].unique()
):

    for top_k in sorted(
        stability_df["Top_K"].unique()
    ):

        subset = stability_df[
            (stability_df["Method"] == method) &
            (stability_df["Top_K"] == top_k)
        ].reset_index(drop=True)

        feature_sets = (
            subset["Feature_Set"]
            .tolist()
        )

        for i in range(
            len(feature_sets)
        ):

            for j in range(
                i + 1,
                len(feature_sets)
            ):

                set_a = feature_sets[i]
                set_b = feature_sets[j]

                intersection = len(
                    set_a.intersection(set_b)
                )

                denominator = (
                    top_k *
                    (
                        N_FEATURES -
                        top_k
                    )
                )

                if denominator == 0:

                    kuncheva = np.nan

                else:

                    kuncheva = (
                        (
                            intersection *
                            N_FEATURES
                        )
                        -
                        (
                            top_k ** 2
                        )
                    ) / denominator

                kuncheva_results.append({

                    "Method": method,

                    "Top_K": top_k,

                    "Run_A": subset.loc[
                        i,
                        "Run"
                    ],

                    "Run_B": subset.loc[
                        j,
                        "Run"
                    ],

                    "Intersection": intersection,

                    "Kuncheva_Index": kuncheva
                })

kuncheva_df = pd.DataFrame(
    kuncheva_results
)

kuncheva_summary = (
    kuncheva_df
    .groupby(
        ["Method", "Top_K"]
    )["Kuncheva_Index"]
    .agg(
        Mean_Kuncheva="mean",
        Std_Kuncheva="std",
        Min_Kuncheva="min",
        Max_Kuncheva="max"
    )
    .reset_index()
)

print("=" * 80)
print("KUNCHEVA INDEX SUMMARY")
print("=" * 80)

display(
    kuncheva_summary.round(4)
)
```

```python
# =============================================================================
# STEP 31: TEST 4 — ANOVA vs DODA FEATURE-SET AGREEMENT
# =============================================================================
# This compares feature-set membership within the SAME repeated-CV run
# and Top-K value.

agreement_results = []

for run in sorted(
    stability_df["Run"].unique()
):

    for top_k in sorted(
        stability_df["Top_K"].unique()
    ):

        anova_rows = stability_df[
            (stability_df["Run"] == run) &
            (stability_df["Method"] == "ANOVA") &
            (stability_df["Top_K"] == top_k)
        ]

        doda_rows = stability_df[
            (stability_df["Run"] == run) &
            (stability_df["Method"] == "DODA") &
            (stability_df["Top_K"] == top_k)
        ]

        if (
            anova_rows.empty or
            doda_rows.empty
        ):
            continue

        anova_set = (
            anova_rows.iloc[0]["Feature_Set"]
        )

        doda_set = (
            doda_rows.iloc[0]["Feature_Set"]
        )

        intersection_size = len(
            anova_set.intersection(
                doda_set
            )
        )

        union_size = len(
            anova_set.union(
                doda_set
            )
        )

        jaccard_agreement = (
            intersection_size / union_size
            if union_size
            else 1.0
        )

        agreement_results.append({

            "Run": run,

            "Top_K": top_k,

            "ANOVA_Features":
                ", ".join(
                    sorted(anova_set)
                ),

            "DODA_Features":
                ", ".join(
                    sorted(doda_set)
                ),

            "Common_Feature_Count":
                intersection_size,

            "ANOVA_Feature_Count":
                len(anova_set),

            "DODA_Feature_Count":
                len(doda_set),

            "Same_Feature_Set":
                anova_set == doda_set,

            "Jaccard_Agreement":
                jaccard_agreement,

            "Features_Added_by_DODA":
                ", ".join(
                    sorted(
                        doda_set -
                        anova_set
                    )
                ),

            "Features_Removed_by_DODA":
                ", ".join(
                    sorted(
                        anova_set -
                        doda_set
                    )
                )
        })

agreement_df = pd.DataFrame(
    agreement_results
)

print("=" * 100)
print("ANOVA vs DODA FEATURE-SET AGREEMENT")
print("=" * 100)

display(agreement_df)
```

```python
# =============================================================================
# STEP 32: SUMMARY OF ANOVA vs DODA AGREEMENT
# =============================================================================

agreement_summary_df = (
    agreement_df
    .groupby("Top_K")
    .agg(
        Same_Feature_Set_Percentage=(
            "Same_Feature_Set",
            lambda x: x.mean() * 100
        ),

        Mean_Jaccard_Agreement=(
            "Jaccard_Agreement",
            "mean"
        ),

        Std_Jaccard_Agreement=(
            "Jaccard_Agreement",
            "std"
        ),

        Mean_Common_Features=(
            "Common_Feature_Count",
            "mean"
        )
    )
    .reset_index()
)

print("=" * 100)
print("FINAL ANOVA vs DODA FEATURE-SET AGREEMENT SUMMARY")
print("=" * 100)

display(
    agreement_summary_df.round(4)
)


# -------------------------------------------------------------------------
# Save stability outputs
# -------------------------------------------------------------------------

os.makedirs(
    "../../../results/breast_cancer/stability",
    exist_ok=True
)

selection_frequency_df.to_csv(
    "../../../results/breast_cancer/stability/feature_selection_frequency.csv",
    index=False
)

jaccard_df.to_csv(
    "../../../results/breast_cancer/stability/jaccard_similarity.csv",
    index=False
)

kuncheva_df.to_csv(
    "../../../results/breast_cancer/stability/kuncheva_pairwise.csv",
    index=False
)

kuncheva_summary.to_csv(
    "../../../results/breast_cancer/stability/kuncheva_summary.csv",
    index=False
)

agreement_df.to_csv(
    "../../../results/breast_cancer/stability/anova_doda_agreement.csv",
    index=False
)

agreement_summary_df.to_csv(
    "../../../results/breast_cancer/stability/anova_doda_agreement_summary.csv",
    index=False
)
```

### 📌 Research Interpretation

The results support the idea that **clinical knowledge can refine statistically selected features** by modifying their ranking toward clinically important variables while generally maintaining competitive predictive performance.

However, the findings should be interpreted within the scope of this experiment:

- The analysis uses **one breast cancer dataset**.
- No statistical significance testing was performed.
- The results do not establish that DODA will generalize to other diseases, datasets, or patient cohorts.
- Further validation on independent datasets is required before making broader claims about DODA.

### ✅ Final Takeaway

**ANOVA provides the statistical foundation, while DODA adds clinical knowledge to refine the feature ranking.**

The results suggest that this combination can preserve strong predictive performance while producing a feature-selection process that is more explicitly informed by clinical priorities.
