```python
# =============================================================================
# STEP 1: LOAD PROCESSED DATASET
# =============================================================================

import pandas as pd
import numpy as np


# Load cleaned dataset
df = pd.read_csv(
    "../../../data/processed/heartdisease_cleaned.csv"
)


print("=" * 70)
print("PROCESSED DATASET")
print("=" * 70)

print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")


display(df.head())
```

```python
# =============================================================================
# STEP 3: VERIFY TARGET DISTRIBUTION
# =============================================================================

print("=" * 70)
print("TARGET DISTRIBUTION")
print("=" * 70)


display(
    y.value_counts()
)


print("\nClass Percentage")


display(
    (y.value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
# =============================================================================
# STEP 8: CLASS DISTRIBUTION AFTER SPLIT
# =============================================================================

print("=" * 70)
print("TRAINING TARGET DISTRIBUTION")
print("=" * 70)


display(
    (y_train.value_counts(normalize=True) * 100)
    .round(2)
)


print("\n" + "=" * 70)
print("TESTING TARGET DISTRIBUTION")
print("=" * 70)


display(
    (y_test.value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
# =============================================================================
# STEP 10: CONVERT SCALED DATA BACK TO DATAFRAME
# =============================================================================


X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns,
    index=X_train.index
)


X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns,
    index=X_test.index
)


print("=" * 70)
print("SCALED TRAIN DATA PREVIEW")
print("=" * 70)


display(
    X_train_scaled.head()
)
```

```python
# =============================================================================
# MULTI-SELECTOR DODA — HEART DISEASE
# STEP 1: IMPORTS
# =============================================================================

import numpy as np
import pandas as pd

# -----------------------------------------------------------------------------
# Scikit-learn
# -----------------------------------------------------------------------------

from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

# -----------------------------------------------------------------------------
# Boruta
# -----------------------------------------------------------------------------

from boruta import BorutaPy

# -----------------------------------------------------------------------------
# mRMR
# -----------------------------------------------------------------------------

from mrmr import mrmr_classif

# -----------------------------------------------------------------------------
# DODA
# -----------------------------------------------------------------------------

from doda import DODASelector
from doda.adapters import SklearnAdapter
from doda.knowledge.providers import JSONProvider
from doda.fusion import RankFusion
from doda.aggregation import MeanAggregator
```

```python
# =============================================================================
# MRMR ADAPTER
# =============================================================================

class MRMRAdapter:

    def __init__(self):
        self.importance = {}

    def fit(self, X, y):

        # Run mRMR for ALL features
        selected_features = mrmr_classif(
            X=X,
            y=y,
            K=X.shape[1]
        )

        # ---------------------------------------------------------------------
        # Convert mRMR ranking into normalized importance scores
        # ---------------------------------------------------------------------

        n_features = len(selected_features)

        scores = np.zeros(len(X.columns))

        feature_to_index = {
            feature: i
            for i, feature in enumerate(X.columns)
        }

        for rank, feature in enumerate(selected_features):

            index = feature_to_index[feature]

            scores[index] = (
                n_features - rank
            ) / n_features

        # ---------------------------------------------------------------------
        # Store importance
        # ---------------------------------------------------------------------

        self.importance = dict(
            zip(
                X.columns,
                scores
            )
        )

        return self

    def get_importance(self):

        return self.importance

    def get_name(self):

        return "MRMR"
```

```python
# =============================================================================
# MULTI-SELECTOR DODA — HEART DISEASE
# STEP 2: CREATE ALL SELECTOR OPERATORS
# =============================================================================

# -----------------------------------------------------------------------------
# 1. ANOVA
# -----------------------------------------------------------------------------

anova_operator = SklearnAdapter(
    SelectKBest(
        score_func=f_classif,
        k="all"
    )
)


# -----------------------------------------------------------------------------
# 2. LASSO
# -----------------------------------------------------------------------------

lasso_model = LogisticRegression(
    penalty="l1",
    solver="liblinear",
    max_iter=5000,
    random_state=42
)

lasso_operator = SklearnAdapter(
    lasso_model
)


# -----------------------------------------------------------------------------
# 3. mRMR
# -----------------------------------------------------------------------------

mrmr_operator = MRMRAdapter()


# -----------------------------------------------------------------------------
# 4. Random Forest
# -----------------------------------------------------------------------------

rf_operator = SklearnAdapter(
    RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight="balanced"
    )
)


# -----------------------------------------------------------------------------
# 5. Boruta
# -----------------------------------------------------------------------------

rf_boruta = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

boruta = BorutaPy(
    estimator=rf_boruta,
    n_estimators="auto",
    max_iter=100,
    perc=100,
    random_state=42,
    verbose=0
)

boruta_operator = SklearnAdapter(
    boruta
)


print("All 5 selector operators created successfully.")
```

```python
# =============================================================================
# MULTI-SELECTOR DODA — HEART DISEASE
# STEP 3: CREATE MEAN AGGREGATOR
# =============================================================================

aggregator = MeanAggregator()

print("MeanAggregator created successfully.")
```

```python
# =============================================================================
# MULTI-SELECTOR DODA — HEART DISEASE
# STEP 4: CLINICAL KNOWLEDGE + FUSION
# =============================================================================

provider = JSONProvider("heartdisease_weights.json")

fusion = RankFusion()

print("Clinical provider and RankFusion created successfully.")
```

```python
# =============================================================================
# MULTI-SELECTOR DODA — HEART DISEASE
# STEP 5: CREATE MULTI-SELECTOR DODA
# =============================================================================

operators = [
    anova_operator,
    lasso_operator,
    mrmr_operator,
    rf_operator,
    boruta_operator
]

selector = DODASelector(
    operators=operators,
    aggregator=aggregator,
    provider=provider,
    fusion=fusion,
    top_k=5
)

print("Multi-selector DODA created successfully.")
print("Number of operators:", len(selector.operators))
```

```python
# =============================================================================
# MULTI-SELECTOR DODA — HEART DISEASE
# STEP 6: FIT DODA + SELECT TOP-5 FEATURES
# =============================================================================

X_train_selected = selector.fit_transform(
    X_train_scaled,
    y_train
)

selected_features = selector.get_selected_features()

print("Selected features:", selected_features)
print("Number of selected features:", len(selected_features))
print("Selected data shape:", X_train_selected.shape)
```

```python
# =============================================================================
# MULTI-SELECTOR DODA — HEART DISEASE
# STEP 7: TRANSFORM TEST DATA
# =============================================================================

X_test_selected = selector.transform(X_test_scaled)

print("Test data shape:", X_test_selected.shape)
print("Selected features:", selector.get_selected_features())
```

```python
# =============================================================================
# MULTI-SELECTOR DODA — HEART DISEASE
# STEP 8: LOGISTIC REGRESSION EVALUATION
# =============================================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

lr_model = LogisticRegression(
    max_iter=5000,
    random_state=42
)

lr_model.fit(X_train_selected, y_train)

y_pred = lr_model.predict(X_test_selected)
y_prob = lr_model.predict_proba(X_test_selected)[:, 1]

lr_results = {
    "Method": "DODA Multi-Selector",
    "Top_K": 5,
    "Model": "Logistic Regression",
    "Selected Features": ", ".join(selector.get_selected_features()),
    "Accuracy": accuracy_score(y_test, y_pred),
    "Precision": precision_score(y_test, y_pred),
    "Recall": recall_score(y_test, y_pred),
    "F1": f1_score(y_test, y_pred),
    "ROC-AUC": roc_auc_score(y_test, y_prob)
}

pd.DataFrame([lr_results])
```

```python
# =============================================================================
# MULTI-SELECTOR DODA — HEART DISEASE
# STEP 9: RANDOM FOREST EVALUATION
# =============================================================================

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

rf_model.fit(X_train_selected, y_train)

y_pred = rf_model.predict(X_test_selected)
y_prob = rf_model.predict_proba(X_test_selected)[:, 1]

rf_results = {
    "Method": "DODA Multi-Selector",
    "Top_K": 5,
    "Model": "Random Forest",
    "Selected Features": ", ".join(selector.get_selected_features()),
    "Accuracy": accuracy_score(y_test, y_pred),
    "Precision": precision_score(y_test, y_pred),
    "Recall": recall_score(y_test, y_pred),
    "F1": f1_score(y_test, y_pred),
    "ROC-AUC": roc_auc_score(y_test, y_prob)
}

pd.DataFrame([rf_results])
```

```python
# =============================================================================
# MULTI-SELECTOR DODA — HEART DISEASE
# STEP 10: XGBOOST EVALUATION
# =============================================================================

from xgboost import XGBClassifier

xgb_model = XGBClassifier(
    n_estimators=200,
    max_depth=3,
    learning_rate=0.05,
    random_state=42,
    eval_metric="logloss"
)

xgb_model.fit(X_train_selected, y_train)

y_pred = xgb_model.predict(X_test_selected)
y_prob = xgb_model.predict_proba(X_test_selected)[:, 1]

xgb_results = {
    "Method": "DODA Multi-Selector",
    "Top_K": 5,
    "Model": "XGBoost",
    "Selected Features": ", ".join(selector.get_selected_features()),
    "Accuracy": accuracy_score(y_test, y_pred),
    "Precision": precision_score(y_test, y_pred),
    "Recall": recall_score(y_test, y_pred),
    "F1": f1_score(y_test, y_pred),
    "ROC-AUC": roc_auc_score(y_test, y_prob)
}

pd.DataFrame([xgb_results])
```

```python
# =============================================================================
# MULTI-SELECTOR DODA — HEART DISEASE
# STEP 11: RUN DODA FOR ALL TOP-K VALUES
# =============================================================================

k_values = [5, 8, 10, 13]

multi_doda_results = {}

for k in k_values:

    print(f"\n{'='*60}")
    print(f"Running Multi-Selector DODA — Top-{k}")
    print(f"{'='*60}")

    selector_k = DODASelector(
        operators=[
            anova_operator,
            lasso_operator,
            mrmr_operator,
            rf_operator,
            boruta_operator
        ],
        aggregator=MeanAggregator(),
        provider=provider,
        fusion=RankFusion(),
        top_k=k
    )

    X_train_k = selector_k.fit_transform(
        X_train_scaled,
        y_train
    )

    X_test_k = selector_k.transform(
        X_test_scaled
    )

    multi_doda_results[k] = {
        "selector": selector_k,
        "X_train": X_train_k,
        "X_test": X_test_k,
        "features": selector_k.get_selected_features()
    }

    print("Selected features:", selector_k.get_selected_features())
    print("Train shape:", X_train_k.shape)
    print("Test shape:", X_test_k.shape)
```

```python
# =============================================================================
# MULTI-SELECTOR DODA — HEART DISEASE
# STEP 12: EVALUATE ALL TOP-K VALUES
# =============================================================================

evaluation_results = []

for k in k_values:

    X_train_k = multi_doda_results[k]["X_train"]
    X_test_k = multi_doda_results[k]["X_test"]
    features_k = multi_doda_results[k]["features"]

    models = {
        "Logistic Regression": LogisticRegression(
            max_iter=5000,
            random_state=42
        ),
        "Random Forest": RandomForestClassifier(
            n_estimators=200,
            random_state=42,
            class_weight="balanced"
        ),
        "XGBoost": XGBClassifier(
            n_estimators=200,
            max_depth=3,
            learning_rate=0.05,
            random_state=42,
            eval_metric="logloss"
        )
    }

    for model_name, model in models.items():

        model.fit(X_train_k, y_train)

        y_pred = model.predict(X_test_k)
        y_prob = model.predict_proba(X_test_k)[:, 1]

        evaluation_results.append({
            "Method": "DODA Multi-Selector",
            "Top_K": k,
            "Model": model_name,
            "Selected Features": ", ".join(features_k),
            "Accuracy": accuracy_score(y_test, y_pred),
            "Precision": precision_score(y_test, y_pred),
            "Recall": recall_score(y_test, y_pred),
            "F1": f1_score(y_test, y_pred),
            "ROC-AUC": roc_auc_score(y_test, y_prob)
        })

doda_multi_results_df = pd.DataFrame(evaluation_results)

doda_multi_results_df
```

```python
# =============================================================================
# MULTI-SELECTOR DODA — HEART DISEASE
# STEP 13: FINAL FEATURE RANKING
# =============================================================================

final_selector = multi_doda_results[13]["selector"]

feature_ranking_df = final_selector.compare_scores()

feature_ranking_df
```
