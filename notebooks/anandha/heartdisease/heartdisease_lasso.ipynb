```python
# =============================================================================
# STEP 1: LOAD PROCESSED DATASET
# =============================================================================

import pandas as pd
import numpy as np


# Load cleaned dataset
df = pd.read_csv(
    "../../../data/processed/heartdisease_cleaned.csv"
)


print("=" * 70)
print("PROCESSED DATASET")
print("=" * 70)

print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")


display(df.head())
```

```python
# =============================================================================
# STEP 2: FEATURE AND TARGET SEPARATION
# =============================================================================

target_column = "target"


X = df.drop(
    target_column,
    axis=1
)


y = df[target_column]


print("=" * 70)
print("FEATURE MATRIX (X)")
print("=" * 70)

print(f"Rows    : {X.shape[0]}")
print(f"Columns : {X.shape[1]}")


print("\n" + "=" * 70)
print("TARGET VARIABLE (y)")
print("=" * 70)

print(f"Rows : {y.shape[0]}")


display(y.head())
```

```python
# =============================================================================
# STEP 3: VERIFY TARGET DISTRIBUTION
# =============================================================================

print("=" * 70)
print("TARGET DISTRIBUTION")
print("=" * 70)


display(
    y.value_counts()
)


print("\nClass Percentage")


display(
    (y.value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
# =============================================================================
# STEP 4: FEATURE TYPE CHECK
# =============================================================================

print("=" * 70)
print("FEATURE DATA TYPES")
print("=" * 70)


display(
    X.dtypes.to_frame(
        name="Data Type"
    )
)
```

```python
# =============================================================================
# STEP 5: MISSING VALUE CHECK
# =============================================================================

print("=" * 70)
print("MISSING VALUES")
print("=" * 70)


missing = pd.DataFrame({
    "Missing Count": X.isnull().sum(),
    "Missing Percentage (%)": (
        X.isnull().mean() * 100
    ).round(2)
})


display(
    missing[missing["Missing Count"] > 0]
)
```

```python
# =============================================================================
# STEP 6: MISSING VALUE IMPUTATION
# =============================================================================

from sklearn.impute import SimpleImputer


imputer = SimpleImputer(
    strategy="median"
)


X_imputed = pd.DataFrame(
    imputer.fit_transform(X),
    columns=X.columns,
    index=X.index
)


print("=" * 70)
print("MISSING VALUE IMPUTATION")
print("=" * 70)

print("Imputation completed successfully")


print("\nRemaining Missing Values:")
print(X_imputed.isnull().sum().sum())
```

```python
# =============================================================================
# STEP 7: TRAIN-TEST SPLIT
# =============================================================================

from sklearn.model_selection import train_test_split


X_train, X_test, y_train, y_test = train_test_split(
    X_imputed,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


print("=" * 70)
print("TRAIN-TEST SPLIT")
print("=" * 70)


print("Training Data")
print("-" * 70)

print(f"X_train shape : {X_train.shape}")
print(f"y_train shape : {y_train.shape}")


print("\nTesting Data")
print("-" * 70)

print(f"X_test shape  : {X_test.shape}")
print(f"y_test shape  : {y_test.shape}")
```

```python
# =============================================================================
# STEP 8: CLASS DISTRIBUTION AFTER SPLIT
# =============================================================================

print("=" * 70)
print("TRAINING TARGET DISTRIBUTION")
print("=" * 70)


display(
    (y_train.value_counts(normalize=True) * 100)
    .round(2)
)


print("\n" + "=" * 70)
print("TESTING TARGET DISTRIBUTION")
print("=" * 70)


display(
    (y_test.value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
# =============================================================================
# STEP 9: FEATURE SCALING
# =============================================================================

from sklearn.preprocessing import StandardScaler


print("=" * 70)
print("FEATURE SCALING USING STANDARD SCALER")
print("=" * 70)


scaler = StandardScaler()


# Fit ONLY on training data
X_train_scaled = scaler.fit_transform(
    X_train
)


# Transform test data
X_test_scaled = scaler.transform(
    X_test
)


print("Scaling completed successfully")


print("\n" + "=" * 70)
print("SHAPE AFTER SCALING")
print("=" * 70)


print(
    f"X_train_scaled shape : {X_train_scaled.shape}"
)

print(
    f"X_test_scaled shape  : {X_test_scaled.shape}"
)
```

```python
# =============================================================================
# STEP 10: CONVERT SCALED DATA BACK TO DATAFRAME
# =============================================================================


X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns,
    index=X_train.index
)


X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns,
    index=X_test.index
)


print("=" * 70)
print("SCALED TRAIN DATA PREVIEW")
print("=" * 70)


display(
    X_train_scaled.head()
)
```

```python
# =============================================================================
# STEP 11: SCALING VERIFICATION
# =============================================================================

print("=" * 70)
print("MEAN AFTER SCALING")
print("=" * 70)


display(
    X_train_scaled.mean().round(2)
)


print("\n" + "=" * 70)
print("STANDARD DEVIATION AFTER SCALING")
print("=" * 70)


display(
    X_train_scaled.std().round(2)
)
```

```python
# =============================================================================
# LASSO BASELINE FEATURE SELECTION (TOP-K = 5,10,15,20)
# =============================================================================

from sklearn.linear_model import LogisticRegression
import pandas as pd
import numpy as np


# Fit LASSO once with all features
# Logistic Regression with L1 penalty = LASSO-style feature selection

lasso_selector = LogisticRegression(
    penalty="l1",
    solver="liblinear",
    C=0.1,
    max_iter=2000,
    class_weight="balanced",
    random_state=42
)


lasso_selector.fit(
    X_train_scaled,
    y_train
)


# =============================================================================
# Extract LASSO coefficients
# =============================================================================

lasso_coefficients = lasso_selector.coef_[0]


# Use absolute coefficient magnitude as feature importance
lasso_scores = pd.DataFrame({

    "Feature": X_train_scaled.columns,

    "LASSO Coefficient": lasso_coefficients,

    "LASSO Score": np.abs(lasso_coefficients)

})


# Sort by absolute coefficient magnitude
lasso_scores = lasso_scores.sort_values(
    by="LASSO Score",
    ascending=False
).reset_index(drop=True)


# =============================================================================
# Top-K values
# =============================================================================

k_values = [5, 8, 10, 13]

lasso_results = {}


for k in k_values:

    top_features = (
        lasso_scores
        .head(k)["Feature"]
        .tolist()
    )


    # Boolean mask for selecting columns
    mask = X_train_scaled.columns.isin(
        top_features
    )


    X_train_selected = X_train_scaled.loc[
        :, mask
    ]

    X_test_selected = X_test_scaled.loc[
        :, mask
    ]


    lasso_results[k] = {

        "features": top_features,

        "X_train": X_train_selected,

        "X_test": X_test_selected,

        "scores": lasso_scores.head(k)

    }


    print("=" * 70)
    print(f"TOP {k} LASSO FEATURES")
    print("=" * 70)

    print(top_features)

    print("\nScores:")
    print(lasso_scores.head(k))
```

```python
# =============================================================================
# BASELINE MODELS
# =============================================================================

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier


models = {

    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        random_state=42,
        eval_metric="logloss",
        scale_pos_weight=(
            (y_train == 0).sum() /
            (y_train == 1).sum()
        )
    )
}
```

```python
# =============================================================================
# MODEL EVALUATION — LASSO
# =============================================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

results_lasso = []


for k in k_values:

    Xtr = lasso_results[k]["X_train"]

    Xts = lasso_results[k]["X_test"]

    features = ", ".join(
        lasso_results[k]["features"]
    )


    for model_name, model in models.items():

        model.fit(
            Xtr,
            y_train
        )


        y_pred = model.predict(
            Xts
        )


        y_prob = model.predict_proba(
            Xts
        )[:, 1]


        results_lasso.append({

            "Method": "LASSO",

            "Top-K": k,

            "Model": model_name,

            "Selected Features": features,

            "Accuracy": accuracy_score(
                y_test,
                y_pred
            ),

            "Precision": precision_score(
                y_test,
                y_pred
            ),

            "Recall": recall_score(
                y_test,
                y_pred
            ),

            "F1 Score": f1_score(
                y_test,
                y_pred
            ),

            "ROC-AUC": roc_auc_score(
                y_test,
                y_prob
            )

        })


lasso_evaluation = pd.DataFrame(
    results_lasso
)

display(lasso_evaluation)
```

```python
lasso_evaluation.to_csv(
    "../../../results/heart_disease/baseline/lasso_baseline_results.csv",
    index=False
)
```

```python
# =============================================================================
# LASSO + DODA FEATURE SELECTION
# =============================================================================

import pandas as pd
from sklearn.linear_model import LogisticRegression


# -------------------------------------------------------------------------
# Top-K values
# -------------------------------------------------------------------------

k_values = [5, 8, 10, 13]


# -------------------------------------------------------------------------
# Clinical knowledge provider
# -------------------------------------------------------------------------

provider = JSONProvider(
    "heartdisease_weights.json"
)


# -------------------------------------------------------------------------
# Fusion method
# -------------------------------------------------------------------------

fusion = RankFusion()


# -------------------------------------------------------------------------
# Store DODA results
# -------------------------------------------------------------------------

results = {}


# =============================================================================
# RUN LASSO + DODA FOR EACH TOP-K
# =============================================================================

for k in k_values:

    print("\n" + "=" * 70)
    print(f"LASSO + DODA — TOP {k} FEATURES")
    print("=" * 70)


    # ---------------------------------------------------------------------
    # LASSO operator
    # ---------------------------------------------------------------------

    lasso = SklearnAdapter(

        LogisticRegression(
            penalty="l1",
            solver="liblinear",
            C=0.1,
            max_iter=2000,
            class_weight="balanced",
            random_state=42
        )

    )


    # ---------------------------------------------------------------------
    # DODA selector
    # ---------------------------------------------------------------------

    selector = DODASelector(

        operators=[lasso],

        provider=provider,

        fusion=fusion,

        top_k=k
    )


    # ---------------------------------------------------------------------
    # Fit DODA on TRAINING data
    # ---------------------------------------------------------------------

    X_train_selected = selector.fit_transform(

        X_train_scaled,

        y_train

    )


    # ---------------------------------------------------------------------
    # Transform TEST data
    # ---------------------------------------------------------------------

    X_test_selected = selector.transform(

        X_test_scaled

    )


    # ---------------------------------------------------------------------
    # Selected features
    # ---------------------------------------------------------------------

    selected_features = (
        selector.get_selected_features()
    )


    print("\nSelected Features:")
    print(selected_features)


    # ---------------------------------------------------------------------
    # Store everything needed for later evaluation
    # ---------------------------------------------------------------------

    results[k] = {

        "selector": selector,

        "X_train": X_train_selected,

        "X_test": X_test_selected,

        "features": selected_features,

        "raw_math_scores":
            selector.raw_math_scores_,

        "math_scores":
            selector.math_scores_,

        "clinical_weights":
            selector.clinical_weights_,

        "final_scores":
            selector.final_scores_
    }


print("\n" + "=" * 70)
print("LASSO + DODA FEATURE SELECTION COMPLETED")
print("=" * 70)
```

```python
# =============================================================================
# INSPECT MATHEMATICAL SCORES
# =============================================================================

k = 10

selector = results[k]["selector"]


# -------------------------------------------------------------------------
# Raw mathematical scores
# -------------------------------------------------------------------------

raw_math_scores = pd.DataFrame(
    selector.raw_math_scores_.items(),
    columns=[
        "Feature",
        "Raw Math Score"
    ]
)


# -------------------------------------------------------------------------
# Normalized mathematical scores
# -------------------------------------------------------------------------

normalized_math_scores = pd.DataFrame(
    selector.math_scores_.items(),
    columns=[
        "Feature",
        "Normalized Math Score"
    ]
)


# -------------------------------------------------------------------------
# Combine both
# -------------------------------------------------------------------------

math_scores = raw_math_scores.merge(
    normalized_math_scores,
    on="Feature"
)


math_scores = math_scores.sort_values(
    by="Normalized Math Score",
    ascending=False
)


print("=" * 70)
print("MATHEMATICAL SCORES")
print("=" * 70)

display(math_scores)
```

```python
# =============================================================================
# INSPECT CLINICAL WEIGHTS
# =============================================================================

clinical_weights = pd.DataFrame(
    selector.clinical_weights_.items(),
    columns=[
        "Feature",
        "Clinical Weight"
    ]
)


clinical_weights = clinical_weights.sort_values(
    by="Clinical Weight",
    ascending=False
)


print("=" * 70)
print("CLINICAL WEIGHTS")
print("=" * 70)

display(clinical_weights)
```

```python
# =============================================================================
# INSPECT FINAL RANK FUSION SCORES
# =============================================================================

final_scores = pd.DataFrame(
    selector.final_scores_.items(),
    columns=[
        "Feature",
        "Final Rank Fusion Score"
    ]
)


final_scores = final_scores.sort_values(
    by="Final Rank Fusion Score",
    ascending=False
)


print("=" * 70)
print("FINAL RANK FUSION SCORES")
print("=" * 70)

display(final_scores)
```

```python
# ============================================================
# LASSO + DODA STABILITY ANALYSIS — ALL TOP-K VALUES
# Heart Disease Dataset
# ============================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression

from doda import DODASelector
from doda.adapters.sklearn_adapter import SklearnAdapter
from doda.knowledge import JSONProvider
from doda.fusion import RankFusion


# ============================================================
# CONFIGURATION
# ============================================================

TOP_K_VALUES = [5, 8, 10, 13]

N_SPLITS = 5
N_REPEATS = 5

RANDOM_STATE = 42

WEIGHTS_FILE = "heartdisease_weights.json"


# ============================================================
# DATA
# ============================================================

# X and y should contain the ORIGINAL Heart Disease data
# BEFORE imputation/scaling.
#
# Example:
#
# X = df.drop(columns=["target"])
# y = df["target"]


# ============================================================
# REPEATED STRATIFIED CROSS-VALIDATION
# ============================================================

cv = RepeatedStratifiedKFold(

    n_splits=N_SPLITS,

    n_repeats=N_REPEATS,

    random_state=RANDOM_STATE

)


# ============================================================
# STORAGE
# ============================================================

all_runs = []

all_jaccard = []

all_frequency = []


# ============================================================
# LOOP THROUGH ALL TOP-K VALUES
# ============================================================

for TOP_K in TOP_K_VALUES:

    print("\n" + "=" * 70)

    print(
        f"LASSO + DODA STABILITY ANALYSIS — TOP-{TOP_K}"
    )

    print("=" * 70)


    selected_feature_sets = []

    run_number = 0


    # ========================================================
    # 25 REPEATED CV RUNS
    # ========================================================

    for train_idx, val_idx in cv.split(X, y):

        run_number += 1


        # ----------------------------------------------------
        # TRAIN / VALIDATION SPLIT
        # ----------------------------------------------------

        X_train = X.iloc[train_idx].copy()

        y_train = y.iloc[train_idx].copy()

        X_val = X.iloc[val_idx].copy()

        y_val = y.iloc[val_idx].copy()


        # ----------------------------------------------------
        # IMPUTATION
        # Fit ONLY on training fold
        # ----------------------------------------------------

        imputer = SimpleImputer(
            strategy="median"
        )


        X_train_imp = pd.DataFrame(

            imputer.fit_transform(
                X_train
            ),

            columns=X_train.columns,

            index=X_train.index

        )


        X_val_imp = pd.DataFrame(

            imputer.transform(
                X_val
            ),

            columns=X_val.columns,

            index=X_val.index

        )


        # ----------------------------------------------------
        # LASSO OPERATOR
        #
        # L1 Logistic Regression = LASSO-style
        # feature selection
        # ----------------------------------------------------

        lasso_operator = SklearnAdapter(

            LogisticRegression(

                penalty="l1",

                solver="liblinear",

                C=0.1,

                max_iter=2000,

                class_weight="balanced",

                random_state=RANDOM_STATE

            )

        )


        # ----------------------------------------------------
        # CLINICAL KNOWLEDGE PROVIDER
        # ----------------------------------------------------

        provider = JSONProvider(
            WEIGHTS_FILE
        )


        # ----------------------------------------------------
        # DODA SELECTOR
        # ----------------------------------------------------

        doda_selector = DODASelector(

            operators=[lasso_operator],

            provider=provider,

            fusion=RankFusion(),

            top_k=TOP_K

        )


        # ----------------------------------------------------
        # FIT DODA ONLY ON TRAINING FOLD
        # ----------------------------------------------------

        doda_selector.fit(

            X_train_imp,

            y_train

        )


        # ----------------------------------------------------
        # GET SELECTED FEATURES
        # ----------------------------------------------------

        selected_features = list(

            doda_selector.get_selected_features()

        )


        # ----------------------------------------------------
        # STORE FEATURE SET
        # ----------------------------------------------------

        selected_feature_sets.append(

            set(selected_features)

        )


        # ----------------------------------------------------
        # STORE RUN INFORMATION
        # ----------------------------------------------------

        all_runs.append({

            "Top_K": TOP_K,

            "Run": run_number,

            "Selected_Features":
                selected_features

        })


        print(

            f"Run {run_number:02d}: "
            f"{selected_features}"

        )


    # ========================================================
    # JACCARD SIMILARITY
    # ========================================================

    jaccard_scores = []


    for i in range(
        len(selected_feature_sets)
    ):

        for j in range(
            i + 1,
            len(selected_feature_sets)
        ):

            set_a = selected_feature_sets[i]

            set_b = selected_feature_sets[j]


            # ------------------------------------------------
            # Intersection
            # ------------------------------------------------

            intersection = len(

                set_a.intersection(
                    set_b
                )

            )


            # ------------------------------------------------
            # Union
            # ------------------------------------------------

            union = len(

                set_a.union(
                    set_b
                )

            )


            # ------------------------------------------------
            # Jaccard
            # ------------------------------------------------

            jaccard = (
                intersection / union
            )


            jaccard_scores.append(
                jaccard
            )


            all_jaccard.append({

                "Top_K": TOP_K,

                "Run_A": i + 1,

                "Run_B": j + 1,

                "Jaccard_Similarity":
                    jaccard

            })


    # ========================================================
    # STABILITY SUMMARY
    # ========================================================

    mean_jaccard = np.mean(
        jaccard_scores
    )

    std_jaccard = np.std(
        jaccard_scores
    )

    min_jaccard = np.min(
        jaccard_scores
    )

    max_jaccard = np.max(
        jaccard_scores
    )


    print("\n" + "-" * 70)

    print(
        f"TOP-{TOP_K} STABILITY SUMMARY"
    )

    print("-" * 70)


    print(
        f"Number of runs:          "
        f"{len(selected_feature_sets)}"
    )


    print(
        f"Number of comparisons:   "
        f"{len(jaccard_scores)}"
    )


    print(
        f"Mean Jaccard Similarity: "
        f"{mean_jaccard:.4f}"
    )


    print(
        f"Standard Deviation:      "
        f"{std_jaccard:.4f}"
    )


    print(
        f"Minimum Jaccard:         "
        f"{min_jaccard:.4f}"
    )


    print(
        f"Maximum Jaccard:         "
        f"{max_jaccard:.4f}"
    )


    # ========================================================
    # FEATURE SELECTION FREQUENCY
    # ========================================================

    feature_counts = {}


    for feature_set in selected_feature_sets:

        for feature in feature_set:

            feature_counts[feature] = (

                feature_counts.get(
                    feature,
                    0
                ) + 1

            )


    # --------------------------------------------------------
    # Store frequency information
    # --------------------------------------------------------

    for feature, count in feature_counts.items():

        all_frequency.append({

            "Top_K": TOP_K,

            "Feature": feature,

            "Selection_Count": count,

            "Selection_Frequency":
                count / len(
                    selected_feature_sets
                )

        })


    # --------------------------------------------------------
    # Display frequency for current Top-K
    # --------------------------------------------------------

    frequency_df_topk = pd.DataFrame(

        [

            {

                "Feature": feature,

                "Selection_Count": count,

                "Selection_Frequency":
                    count /
                    len(selected_feature_sets)

            }

            for feature, count
            in feature_counts.items()

        ]

    ).sort_values(

        "Selection_Frequency",

        ascending=False

    )


    print(
        "\nFEATURE SELECTION FREQUENCY"
    )


    print(

        frequency_df_topk.to_string(
            index=False
        )

    )


# ============================================================
# CREATE FINAL DATAFRAMES
# ============================================================

runs_df = pd.DataFrame(
    all_runs
)

jaccard_df = pd.DataFrame(
    all_jaccard
)

frequency_df = pd.DataFrame(
    all_frequency
)


# ============================================================
# FINAL JACCARD SUMMARY
# ============================================================

summary_rows = []


for TOP_K in TOP_K_VALUES:

    subset = jaccard_df[

        jaccard_df["Top_K"] == TOP_K

    ]["Jaccard_Similarity"]


    summary_rows.append({

        "Top_K": TOP_K,

        "Mean_Jaccard":
            subset.mean(),

        "Std_Jaccard":
            subset.std(),

        "Min_Jaccard":
            subset.min(),

        "Max_Jaccard":
            subset.max()

    })


summary_df = pd.DataFrame(
    summary_rows
)


# ============================================================
# DISPLAY FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)

print(
    "LASSO + DODA STABILITY — ALL TOP-K VALUES"
)

print("=" * 70)


print(

    summary_df.to_string(

        index=False,

        float_format=lambda x:
            f"{x:.4f}"

    )

)


# ============================================================
# SAVE RESULTS
# ============================================================

runs_df.to_csv(

    "../../../results/heart_disease/"
    "lasso_doda_stability_all_topk_runs.csv",

    index=False

)


jaccard_df.to_csv(

    "../../../results/heart_disease/"
    "lasso_doda_stability_all_topk_jaccard.csv",

    index=False

)


frequency_df.to_csv(

    "../../../results/heart_disease/"
    "lasso_doda_stability_all_topk_frequency.csv",

    index=False

)


summary_df.to_csv(

    "../../../results/heart_disease/"
    "lasso_doda_stability_all_topk_summary.csv",

    index=False

)


print("\nFiles saved:")

print(
    "1. lasso_doda_stability_all_topk_runs.csv"
)

print(
    "2. lasso_doda_stability_all_topk_jaccard.csv"
)

print(
    "3. lasso_doda_stability_all_topk_frequency.csv"
)

print(
    "4. lasso_doda_stability_all_topk_summary.csv"
)
```

```python
# =============================================================================
# REPEATED STRATIFIED CROSS-VALIDATION
# LASSO VS DODA-LASSO WITH RANK FUSION
# =============================================================================

import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)


from doda import DODASelector
from doda.adapters.sklearn_adapter import SklearnAdapter
from doda.knowledge import JSONProvider
from doda.fusion import RankFusion


# =============================================================================
# CONFIGURATION
# =============================================================================

TOP_K_VALUES = [5, 8, 10, 13]

N_SPLITS = 5
N_REPEATS = 5
RANDOM_STATE = 42

target_column = "target"

X = df.drop(columns=[target_column])
y = df[target_column]

feature_names = X.columns.tolist()


# =============================================================================
# MODELS
# =============================================================================

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=RANDOM_STATE
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),

    "XGBoost": XGBClassifier(
        n_estimators=300,
        max_depth=3,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=RANDOM_STATE,
        n_jobs=-1
    )
}


# =============================================================================
# CLINICAL KNOWLEDGE PROVIDER
# =============================================================================

provider = JSONProvider(
    "heartdisease_weights.json"
)


# =============================================================================
# REPEATED STRATIFIED CROSS-VALIDATION
# =============================================================================

cv = RepeatedStratifiedKFold(
    n_splits=N_SPLITS,
    n_repeats=N_REPEATS,
    random_state=RANDOM_STATE
)


# =============================================================================
# RESULTS STORAGE
# =============================================================================

results = []


# =============================================================================
# CROSS-VALIDATION LOOP
# =============================================================================

for run_number, (train_idx, test_idx) in enumerate(
    cv.split(X, y),
    start=1
):

    print("\n" + "=" * 80)
    print(f"REPEATED STRATIFIED CV RUN {run_number}")
    print("=" * 80)

    X_train = X.iloc[train_idx].copy()
    X_test = X.iloc[test_idx].copy()

    y_train = y.iloc[train_idx].copy()
    y_test = y.iloc[test_idx].copy()


    # =========================================================================
    # IMPUTATION
    # =========================================================================

    imputer = SimpleImputer(
        strategy="median"
    )

    X_train_imp_array = imputer.fit_transform(
        X_train
    )

    X_test_imp_array = imputer.transform(
        X_test
    )

    X_train_imp = pd.DataFrame(
        X_train_imp_array,
        columns=feature_names,
        index=X_train.index
    )

    X_test_imp = pd.DataFrame(
        X_test_imp_array,
        columns=feature_names,
        index=X_test.index
    )


    # =========================================================================
    # FEATURE SELECTION FOR EACH TOP-K VALUE
    # =========================================================================

    for top_k in TOP_K_VALUES:

        print("\n" + "-" * 80)
        print(f"TOP-K = {top_k}")
        print("-" * 80)


        # =====================================================================
        # BASELINE LASSO FEATURE SELECTION
        # =====================================================================

        # Standardization is performed only using training data.
        # This is important because LASSO is scale-sensitive.

        lasso_scaler = StandardScaler()

        X_train_lasso_scaled = lasso_scaler.fit_transform(
            X_train_imp
        )

        X_test_lasso_scaled = lasso_scaler.transform(
            X_test_imp
        )

        lasso_model = LogisticRegression(
            penalty="l1",
            solver="liblinear",
            C=0.1,
            max_iter=2000,
            random_state=RANDOM_STATE
        )

        lasso_model.fit(
            X_train_lasso_scaled,
            y_train
        )

        lasso_coefficients = np.abs(
            lasso_model.coef_[0]
        )

        lasso_ranked_indices = np.argsort(
            lasso_coefficients
        )[::-1]

        lasso_selected_indices = (
            lasso_ranked_indices[:top_k]
        )

        lasso_features = [
            feature_names[i]
            for i in lasso_selected_indices
        ]

        print("LASSO-selected features:")
        print(lasso_features)


        # =====================================================================
        # DODA-LASSO FEATURE SELECTION
        # =====================================================================

        # DODA receives the L1 logistic-regression estimator directly.
        # Do not wrap it inside SelectFromModel because the DODA adapter
        # extracts the coef_ attribute from the estimator.

        doda_lasso_scaler = StandardScaler()

        X_train_doda_scaled_array = (
            doda_lasso_scaler.fit_transform(
                X_train_imp
            )
        )

        X_test_doda_scaled_array = (
            doda_lasso_scaler.transform(
                X_test_imp
            )
        )

        X_train_doda_scaled = pd.DataFrame(
            X_train_doda_scaled_array,
            columns=feature_names,
            index=X_train_imp.index
        )

        X_test_doda_scaled = pd.DataFrame(
            X_test_doda_scaled_array,
            columns=feature_names,
            index=X_test_imp.index
        )

        doda_lasso_estimator = LogisticRegression(
            penalty="l1",
            solver="liblinear",
            C=0.1,
            max_iter=2000,
            random_state=RANDOM_STATE
        )

        lasso_operator = SklearnAdapter(
            doda_lasso_estimator
        )

        doda_selector = DODASelector(
            operators=[lasso_operator],
            provider=provider,
            fusion=RankFusion(),
            top_k=top_k
        )

        doda_selector.fit(
            X_train_doda_scaled,
            y_train
        )

        doda_features = (
            doda_selector.get_selected_features()
        )

        print("DODA-selected features:")
        print(doda_features)


        # =====================================================================
        # PREPARE SELECTED FEATURE MATRICES
        # =====================================================================

        X_train_lasso_selected = (
            X_train_imp[lasso_features]
        )

        X_test_lasso_selected = (
            X_test_imp[lasso_features]
        )

        X_train_doda_selected = (
            X_train_imp[doda_features]
        )

        X_test_doda_selected = (
            X_test_imp[doda_features]
        )


        # =====================================================================
        # MODEL EVALUATION
        # =====================================================================

        selected_feature_sets = {
            "LASSO": (
                X_train_lasso_selected,
                X_test_lasso_selected,
                lasso_features
            ),

            "DODA": (
                X_train_doda_selected,
                X_test_doda_selected,
                doda_features
            )
        }


        for method, (
            X_train_selected,
            X_test_selected,
            selected_features
        ) in selected_feature_sets.items():

            for model_name, model_template in models.items():

                model = clone(
                    model_template
                )

                X_train_model = X_train_selected.copy()
                X_test_model = X_test_selected.copy()


                # -------------------------------------------------------------
                # Scale only Logistic Regression
                # -------------------------------------------------------------

                if model_name == "Logistic Regression":

                    model_scaler = StandardScaler()

                    X_train_model = (
                        model_scaler.fit_transform(
                            X_train_model
                        )
                    )

                    X_test_model = (
                        model_scaler.transform(
                            X_test_model
                        )
                    )


                # -------------------------------------------------------------
                # Train model
                # -------------------------------------------------------------

                model.fit(
                    X_train_model,
                    y_train
                )

                y_pred = model.predict(
                    X_test_model
                )

                y_prob = model.predict_proba(
                    X_test_model
                )[:, 1]


                # -------------------------------------------------------------
                # Calculate metrics
                # -------------------------------------------------------------

                accuracy = accuracy_score(
                    y_test,
                    y_pred
                )

                precision = precision_score(
                    y_test,
                    y_pred,
                    zero_division=0
                )

                recall = recall_score(
                    y_test,
                    y_pred,
                    zero_division=0
                )

                f1 = f1_score(
                    y_test,
                    y_pred,
                    zero_division=0
                )

                roc_auc = roc_auc_score(
                    y_test,
                    y_prob
                )


                # -------------------------------------------------------------
                # Store results
                # -------------------------------------------------------------

                results.append({
                    "Run": run_number,
                    "Method": method,
                    "Top_K": top_k,
                    "Model": model_name,
                    "Accuracy": accuracy,
                    "Precision": precision,
                    "Recall": recall,
                    "F1": f1,
                    "ROC_AUC": roc_auc,
                    "Selected_Features": selected_features
                })


# =============================================================================
# RESULTS DATAFRAME
# =============================================================================

results_df = pd.DataFrame(
    results
)

print("\n" + "=" * 80)
print("REPEATED STRATIFIED CV RESULTS")
print("=" * 80)

display(
    results_df.head()
)

print(
    "Total evaluation rows:",
    len(results_df)
)
```

```python
# =============================================================================
# PREPARE FEATURE SETS FOR STABILITY ANALYSIS
# =============================================================================

import ast
import numpy as np
import pandas as pd
from itertools import combinations


def convert_to_set(features):
    """
    Convert Selected_Features into a Python set.
    Handles lists, tuples, NumPy arrays, and strings.
    """

    if isinstance(features, str):

        # Handle string representation of a list
        if features.startswith("[") and features.endswith("]"):
            try:
                features = ast.literal_eval(features)
            except Exception:
                features = features.split(",")

        else:
            features = features.split(",")

    return {
        str(feature).strip().strip("'").strip('"')
        for feature in features
        if str(feature).strip()
    }


# Use only one model because feature selection is independent of evaluation model
stability_df = results_df[
    results_df["Model"] == "Logistic Regression"
].copy()

stability_df["Feature_Set"] = stability_df["Selected_Features"].apply(
    convert_to_set
)

print("Rows used for stability analysis:", len(stability_df))
print(stability_df[["Run", "Method", "Top_K", "Feature_Set"]].head())
```

```python
# =============================================================================
# TEST 1: FEATURE SELECTION FREQUENCY
# =============================================================================

selection_frequency = []

for method in stability_df["Method"].unique():

    for top_k in sorted(stability_df["Top_K"].unique()):

        subset = stability_df[
            (stability_df["Method"] == method) &
            (stability_df["Top_K"] == top_k)
        ]

        total_runs = len(subset)

        all_features = sorted(
            set().union(*subset["Feature_Set"])
        )

        for feature in all_features:

            selected_count = sum(
                feature in feature_set
                for feature_set in subset["Feature_Set"]
            )

            selection_frequency.append({
                "Method": method,
                "Top_K": top_k,
                "Feature": feature,
                "Selected_Count": selected_count,
                "Total_Runs": total_runs,
                "Selection_Frequency_Percentage": (
                    selected_count / total_runs
                ) * 100
            })

selection_frequency_df = pd.DataFrame(selection_frequency)

selection_frequency_df = selection_frequency_df.sort_values(
    by=[
        "Method",
        "Top_K",
        "Selection_Frequency_Percentage"
    ],
    ascending=[True, True, False]
).reset_index(drop=True)

display(selection_frequency_df)
```

```python
# =============================================================================
# TEST 2: PAIRWISE JACCARD SIMILARITY
# =============================================================================

jaccard_results = []

for method in stability_df["Method"].unique():

    for top_k in sorted(stability_df["Top_K"].unique()):

        subset = stability_df[
            (stability_df["Method"] == method) &
            (stability_df["Top_K"] == top_k)
        ].reset_index(drop=True)

        feature_sets = subset["Feature_Set"].tolist()

        pairwise_scores = []

        for set_a, set_b in combinations(feature_sets, 2):

            intersection_size = len(set_a.intersection(set_b))
            union_size = len(set_a.union(set_b))

            if union_size == 0:
                jaccard_score = 1.0
            else:
                jaccard_score = intersection_size / union_size

            pairwise_scores.append(jaccard_score)

        jaccard_results.append({
            "Method": method,
            "Top_K": top_k,
            "Mean_Jaccard": np.mean(pairwise_scores),
            "Std_Jaccard": np.std(pairwise_scores, ddof=1),
            "Minimum_Jaccard": np.min(pairwise_scores),
            "Maximum_Jaccard": np.max(pairwise_scores)
        })

jaccard_df = pd.DataFrame(jaccard_results)

display(jaccard_df)
```

```python
# =============================================================================
# TEST 3: KUNCHEVA INDEX
# =============================================================================

TOTAL_FEATURES = 13

kuncheva_results = []

for method in stability_df["Method"].unique():

    for top_k in sorted(stability_df["Top_K"].unique()):

        subset = stability_df[
            (stability_df["Method"] == method) &
            (stability_df["Top_K"] == top_k)
        ].reset_index(drop=True)

        feature_sets = subset["Feature_Set"].tolist()

        pairwise_scores = []

        for set_a, set_b in combinations(feature_sets, 2):

            k_a = len(set_a)
            k_b = len(set_b)

            # Skip pairs with different numbers of selected features
            if k_a != k_b:
                continue

            k = k_a

            # Kuncheva index is undefined for empty or complete feature sets
            if k == 0 or k >= TOTAL_FEATURES:
                continue

            common_features = len(set_a.intersection(set_b))

            expected_overlap = (k * k) / TOTAL_FEATURES

            denominator = k - expected_overlap

            if denominator == 0:
                continue

            kuncheva_score = (
                common_features - expected_overlap
            ) / denominator

            pairwise_scores.append(kuncheva_score)

        if len(pairwise_scores) == 0:

            kuncheva_results.append({
                "Method": method,
                "Top_K": top_k,
                "Mean_Kuncheva": np.nan,
                "Std_Kuncheva": np.nan,
                "Minimum_Kuncheva": np.nan,
                "Maximum_Kuncheva": np.nan,
                "Valid_Pair_Count": 0
            })

        else:

            kuncheva_results.append({
                "Method": method,
                "Top_K": top_k,
                "Mean_Kuncheva": np.mean(pairwise_scores),
                "Std_Kuncheva": (
                    np.std(pairwise_scores, ddof=1)
                    if len(pairwise_scores) > 1
                    else 0.0
                ),
                "Minimum_Kuncheva": np.min(pairwise_scores),
                "Maximum_Kuncheva": np.max(pairwise_scores),
                "Valid_Pair_Count": len(pairwise_scores)
            })

kuncheva_df = pd.DataFrame(kuncheva_results)

display(kuncheva_df)
```

```python
# =============================================================================
# TEST 4: LASSO VS DODA FEATURE-SET AGREEMENT
# =============================================================================

agreement_results = []

for run in sorted(stability_df["Run"].unique()):

    for top_k in sorted(stability_df["Top_K"].unique()):

        lasso_rows = stability_df[
            (stability_df["Run"] == run) &
            (stability_df["Method"] == "LASSO") &
            (stability_df["Top_K"] == top_k)
        ]

        doda_rows = stability_df[
            (stability_df["Run"] == run) &
            (stability_df["Method"] == "DODA") &
            (stability_df["Top_K"] == top_k)
        ]

        if lasso_rows.empty or doda_rows.empty:
            continue

        lasso_set = lasso_rows.iloc[0]["Feature_Set"]
        doda_set = doda_rows.iloc[0]["Feature_Set"]

        intersection_size = len(lasso_set.intersection(doda_set))
        union_size = len(lasso_set.union(doda_set))

        if union_size == 0:
            jaccard_agreement = 1.0
        else:
            jaccard_agreement = intersection_size / union_size

        agreement_results.append({
            "Run": run,
            "Top_K": top_k,
            "LASSO_Features": ", ".join(sorted(lasso_set)),
            "DODA_Features": ", ".join(sorted(doda_set)),
            "Common_Feature_Count": intersection_size,
            "LASSO_Feature_Count": len(lasso_set),
            "DODA_Feature_Count": len(doda_set),
            "Same_Feature_Set": lasso_set == doda_set,
            "Jaccard_Agreement": jaccard_agreement
        })

agreement_df = pd.DataFrame(agreement_results)

display(agreement_df)
```

```python
# =============================================================================
# SUMMARY OF LASSO VS DODA AGREEMENT
# =============================================================================

agreement_summary_df = (
    agreement_df
    .groupby("Top_K")
    .agg(
        Same_Feature_Set_Percentage=(
            "Same_Feature_Set",
            lambda x: x.mean() * 100
        ),
        Mean_Jaccard_Agreement=(
            "Jaccard_Agreement",
            "mean"
        ),
        Std_Jaccard_Agreement=(
            "Jaccard_Agreement",
            "std"
        ),
        Mean_Common_Features=(
            "Common_Feature_Count",
            "mean"
        )
    )
    .reset_index()
)

display(agreement_summary_df)
```

## LASSO Feature Selection — Heart Disease

### Overall Interpretation

The LASSO experiment evaluated conventional LASSO feature selection against **LASSO + Rank Fusion DODA**, using Logistic Regression, Random Forest, and XGBoost across Top-5, Top-8, Top-10, and Top-13 feature subsets.

Unlike the earlier Hadamard fusion approach, the current DODA implementation uses **Rank Fusion**, combining the LASSO-derived feature ranking with clinically informed feature ranking. The results show that Rank Fusion substantially changes the selected feature composition at lower Top-K values while retaining a clinically consistent core of features.

The results demonstrate that DODA does **not uniformly improve every predictive metric or every model**. However, it provides strong feature-selection stability and, particularly at Top-8 and Top-10, maintains or improves predictive performance for several models.

---

## 1. LASSO Baseline Feature Selection

The LASSO baseline identified a consistent core of heart-disease-related features.

At Top-5, the selected features were:

`ca, thal, cp, exang, sex`

At Top-8, the feature set expanded to:

`ca, thal, cp, exang, sex, thalach, slope, oldpeak`

At Top-10, `restecg` and `chol` were additionally included, while Top-13 retained all available features.

This indicates that LASSO consistently prioritised `ca`, `thal`, `cp`, and `exang` among the most important predictors.

---

## 2. Effect of Rank Fusion DODA on Feature Selection

After applying Rank Fusion DODA, the Top-5 feature set changed from:

**LASSO:**
`ca, thal, cp, exang, sex`

to:

**LASSO + DODA:**
`ca, thal, cp, exang, oldpeak`

Thus, DODA replaced `sex` with `oldpeak` at Top-5.

This is important because `oldpeak` was strongly retained by the DODA ranking: it was selected in **23/25 runs (92%)** at Top-5.

At Top-8, DODA selected:

`ca, thal, cp, exang, oldpeak, slope, thalach, age`

Here, `sex` was replaced by `age` compared with the LASSO baseline.

At Top-10, DODA selected:

`ca, thal, cp, exang, oldpeak, slope, thalach, sex, age, trestbps`

Compared with the LASSO Top-10 subset, DODA introduced `age` and `trestbps` while removing `restecg` and `chol`.

Therefore, Rank Fusion does not simply reproduce the LASSO ranking. It modifies the feature composition by incorporating the clinical ranking while preserving several of LASSO's strongest features.

---

## 3. Top-5 Predictive Performance

At Top-5, DODA produced mixed changes across the three models.

### Logistic Regression

DODA increased:

* Accuracy: **81.97% → 85.25%**
* Recall: **85.71% → 89.29%**
* F1-score: **81.36% → 84.75%**
* ROC-AUC: **0.8983 → 0.9405**

This represents an improvement across all reported metrics.

### Random Forest

DODA decreased accuracy from **81.97% to 72.13%** and F1-score from **80.70% to 72.13%**.

However, ROC-AUC decreased only slightly from **0.9048 to 0.8799**.

### XGBoost

DODA produced a small reduction in accuracy (**83.61% → 78.69%**) and F1-score (**82.76% → 77.19%**), while ROC-AUC decreased from **0.9242 to 0.8793**.

Therefore, at Top-5, the effect of DODA was strongly model-dependent. Logistic Regression benefited substantially, whereas the tree-based models performed better with the original LASSO feature subset.

---

## 4. Top-8 Predictive Performance

At Top-8, DODA selected:

`ca, thal, cp, exang, oldpeak, slope, thalach, sex`

The DODA subset produced:

### Logistic Regression

* Accuracy: **85.25% → 85.25%**
* F1-score: **84.75% → 84.75%**
* ROC-AUC: **0.9318 → 0.9351**

Thus, predictive performance was essentially preserved, with a small improvement in ROC-AUC.

### Random Forest

* Accuracy remained **83.61%**
* F1-score increased from **82.76% to 83.87%**
* ROC-AUC decreased slightly from **0.9432 to 0.9389**

This indicates broadly comparable performance.

### XGBoost

DODA increased:

* Accuracy: **81.97% → 85.25%**
* F1-score: **81.36% → 84.21%**
* ROC-AUC: **0.9307 → 0.9253**

The slight decrease in ROC-AUC occurred alongside improvements in accuracy and F1-score.

Overall, Top-8 provides evidence that DODA can alter the feature set while maintaining or improving several predictive measures.

---

## 5. Top-10 Predictive Performance

At Top-10, DODA selected:

`ca, thal, cp, exang, oldpeak, slope, thalach, sex, age, trestbps`

This produced particularly consistent results.

### Logistic Regression

* Accuracy: **85.25% → 86.89%**
* Recall: **92.86% → 92.86%**
* F1-score: **85.25% → 86.67%**
* ROC-AUC: **0.9329 → 0.9448**

Thus, DODA improved all major predictive measures except recall, which remained unchanged.

### Random Forest

* Accuracy: **88.52% → 86.89%**
* F1-score: **88.14% → 86.67%**
* ROC-AUC: **0.9416 → 0.9481**

Although accuracy and F1-score decreased, ROC-AUC improved.

### XGBoost

* Accuracy: **83.61% → 85.25%**
* Recall: **89.29% → 96.43%**
* F1-score: **83.33% → 85.71%**
* ROC-AUC: **0.9134 → 0.9416**

The most notable improvement at Top-10 was for XGBoost, particularly the increase in recall from **89.29% to 96.43%** and ROC-AUC from **0.9134 to 0.9416**.

---

## 6. Top-13

At Top-13, all available features are retained, so the feature-selection difference between LASSO and DODA disappears.

The two approaches therefore operate on the same complete feature set.

This provides a useful reference point because any differences at Top-5, Top-8, and Top-10 can be attributed to the different feature-selection strategies rather than different numbers of available predictors.

---

## 7. Cross-Validation Results

The repeated 5-fold results provide a more robust view than the single evaluation results.

For LASSO, mean ROC-AUC values ranged from:

* **0.8941–0.9116** for Logistic Regression
* **0.8671–0.9104** for Random Forest
* **0.8634–0.8931** for XGBoost

For LASSO + DODA, mean ROC-AUC values ranged from:

* **0.8954–0.9116** for Logistic Regression
* **0.8707–0.9085** for Random Forest
* **0.8723–0.8964** for XGBoost

The cross-validation results therefore show that DODA generally maintained **comparable discriminative performance** to LASSO rather than producing a universal performance increase.

The strongest DODA result in the repeated evaluation occurs at Top-10 for Logistic Regression, where ROC-AUC is **0.9099**, compared with **0.9099** for LASSO, while accuracy increases from **83.82% to 84.15%**.

For XGBoost at Top-10, DODA increases mean accuracy from **80.53% to 81.84%**, recall from **74.07% to 77.65%**, and F1-score from **77.70% to 79.53%**.

---

## 8. Feature-Selection Stability

Feature-selection stability is one of the strongest findings of the LASSO-DODA experiment.

| Top-K | Mean Jaccard |         SD | Interpretation      |
| ----: | -----------: | ---------: | ------------------- |

The Top-8 DODA feature set was selected identically across all 25 runs.

All eight features:

`ca, thal, cp, exang, oldpeak, slope, thalach, age`

had a selection frequency of **100%**.

At Top-10, most features were also highly stable:

* `ca`: 100%

Only `restecg`, which appeared occasionally during the repeated Top-10 selections, had a low selection frequency of **16%**.

This indicates that the DODA ranking has a particularly stable core of eight features.

---

## 9. Overall Interpretation

The Heart Disease LASSO experiment demonstrates that **Rank Fusion DODA modifies the LASSO feature ranking while preserving a strong and highly reproducible core of features**.

The most consistent features across the DODA rankings were:

`ca, thal, cp, exang, oldpeak, slope, thalach, age`

The Top-8 configuration is particularly notable because all eight features were selected in **25/25 repeated runs**, producing a mean Jaccard similarity of **1.0000**.

Predictively, DODA does not consistently outperform LASSO across every model and Top-K configuration. Instead, its effect depends on the downstream model and number of selected features. Logistic Regression generally showed stable or improved performance, while Random Forest and XGBoost showed both improvements and reductions depending on Top-K.

The Top-10 configuration provides particularly useful evidence for DODA, with improvements in several classification metrics and a substantial increase in XGBoost recall. At the same time, the cross-validation results demonstrate that these improvements should be interpreted as **model- and Top-K-specific rather than as a universal performance gain**.

Overall, the LASSO experiment supports the role of Rank Fusion DODA as a **clinically guided re-ranking mechanism that can produce highly stable feature subsets while maintaining broadly comparable predictive performance**. Its contribution is therefore not limited to maximizing predictive accuracy, but also includes the integration of clinical information and the generation of reproducible feature rankings.
