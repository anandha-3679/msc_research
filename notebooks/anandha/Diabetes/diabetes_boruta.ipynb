`````python
# =============================================================================
# BORUTA BASELINE FEATURE SELECTION (TOP-K = 5,10,15,20)
# =============================================================================

from boruta import BorutaPy
from sklearn.ensemble import RandomForestClassifier
import pandas as pd
import numpy as np


# =============================================================================
# BORUTA FEATURE SELECTION
# =============================================================================

# Random Forest used internally by Boruta
rf_boruta = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)


# Initialize Boruta
boruta_selector = BorutaPy(
    estimator=rf_boruta,

    n_estimators="auto",

    verbose=0,

    random_state=42,

    max_iter=100,

    perc=100
)


# -------------------------------------------------------------------------
# Fit Boruta ONLY on training data
# -------------------------------------------------------------------------

boruta_selector.fit(
    X_train_scaled.values,
    y_train.values
)


# =============================================================================
# CREATE BORUTA RANKING
# =============================================================================

boruta_scores = pd.DataFrame({

    "Feature": X_train_scaled.columns,

    "Boruta Rank": boruta_selector.ranking_,

    "Confirmed": boruta_selector.support_,

    "Tentative": boruta_selector.support_weak_

})


# -------------------------------------------------------------------------
# Sort by Boruta rank
# -------------------------------------------------------------------------

boruta_scores = boruta_scores.sort_values(
    by=["Boruta Rank", "Feature"]
).reset_index(drop=True)


print("=" * 70)
print("BORUTA FEATURE RANKING")
print("=" * 70)

print(boruta_scores)


# =============================================================================
# TOP-K FEATURE SELECTION
# =============================================================================

k_values = [5, 10, 15, 20]

boruta_results = {}


for k in k_values:

    # ---------------------------------------------------------------------
    # Select top-k features according to Boruta ranking
    # ---------------------------------------------------------------------

    top_features = (
        boruta_scores
        .head(k)["Feature"]
        .tolist()
    )


    # ---------------------------------------------------------------------
    # Boolean mask for selecting columns
    # ---------------------------------------------------------------------

    mask = X_train_scaled.columns.isin(
        top_features
    )


    # ---------------------------------------------------------------------
    # Select training and testing features
    # ---------------------------------------------------------------------

    X_train_selected = X_train_scaled.loc[
        :, mask
    ]

    X_test_selected = X_test_scaled.loc[
        :, mask
    ]


    # ---------------------------------------------------------------------
    # Store results
    # ---------------------------------------------------------------------

    boruta_results[k] = {

        "features": top_features,

        "X_train": X_train_selected,

        "X_test": X_test_selected,

        "scores": boruta_scores.head(k)

    }


    # ---------------------------------------------------------------------
    # Display
    # ---------------------------------------------------------------------

    print("=" * 70)
    print(f"TOP {k} BORUTA FEATURES")
    print("=" * 70)

    print(top_features)

    print("\nBoruta Ranking:")
    print(
        boruta_scores.head(k)
    )
`````

`````python
# =============================================================================
# BASELINE MODELS
# =============================================================================

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier


models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        random_state=42,

        eval_metric="logloss",
        scale_pos_weight=(y_train == 0).sum() / (y_train == 1).sum()
    )
}
`````

`````python
# =============================================================================
# MODEL EVALUATION
# =============================================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)


results = []


for k in k_values:

    Xtr = boruta_results[k]["X_train"]

    Xts = boruta_results[k]["X_test"]

    features = ", ".join(
        boruta_results[k]["features"]
    )


    # ---------------------------------------------------------------------
    # Evaluate every model
    # ---------------------------------------------------------------------

    for model_name, model in models.items():

        # Fit model
        model.fit(
            Xtr,
            y_train
        )


        # Predictions
        y_pred = model.predict(
            Xts
        )


        # Probability predictions
        y_prob = model.predict_proba(
            Xts
        )[:, 1]


        # Store metrics
        results.append({

            "Method": "Boruta",

            "Top-K": k,

            "Model": model_name,

            "Selected Features": features,

            "Accuracy": accuracy_score(
                y_test,
                y_pred
            ),

            "Precision": precision_score(
                y_test,
                y_pred,
                zero_division=0
            ),

            "Recall": recall_score(
                y_test,
                y_pred,
                zero_division=0
            ),

            "F1 Score": f1_score(
                y_test,
                y_pred,
                zero_division=0
            ),

            "ROC-AUC": roc_auc_score(
                y_test,
                y_prob
            )

        })


# =============================================================================
# FINAL BORUTA RESULTS TABLE
# =============================================================================

boruta_results_df = pd.DataFrame(
    results
)


display(
    boruta_results_df
)
`````

`````python
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import f_classif

from doda import DODASelector

from doda.adapters import SklearnAdapter

from doda.knowledge.providers import JSONProvider

from doda.fusion import HadamardFusion
`````

`````python
# =============================================================================
# BORUTA + DODA FEATURE SELECTION
# =============================================================================

import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from boruta import BorutaPy


# -------------------------------------------------------------------------
# Top-K values
# -------------------------------------------------------------------------

k_values = [5, 10, 15, 20]


# -------------------------------------------------------------------------
# Clinical knowledge provider
# -------------------------------------------------------------------------

provider = JSONProvider(
    "clinical_weights.json"
)


# -------------------------------------------------------------------------
# Fusion method
# -------------------------------------------------------------------------

fusion = HadamardFusion()


# -------------------------------------------------------------------------
# Store DODA results
# -------------------------------------------------------------------------

doda_results = {}


# =============================================================================
# RUN BORUTA + DODA FOR EACH TOP-K
# =============================================================================

for k in k_values:

    print("\n" + "=" * 70)
    print(f"BORUTA + DODA — TOP {k} FEATURES")
    print("=" * 70)


    # ---------------------------------------------------------------------
    # Random Forest estimator used by Boruta
    # ---------------------------------------------------------------------

    rf_boruta = RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )


    # ---------------------------------------------------------------------
    # Boruta operator
    # ---------------------------------------------------------------------

    boruta = BorutaPy(
        estimator=rf_boruta,
        n_estimators="auto",
        max_iter=100,
        perc=100,
        random_state=42,
        verbose=0
    )


    # ---------------------------------------------------------------------
    # Wrap Boruta as a DODA operator
    # ---------------------------------------------------------------------

    boruta_operator = SklearnAdapter(
        boruta
    )


    # ---------------------------------------------------------------------
    # DODA selector
    # ---------------------------------------------------------------------

    selector = DODASelector(
        operators=[boruta_operator],
        provider=provider,
        fusion=fusion,
        top_k=k
    )


    # ---------------------------------------------------------------------
    # Fit DODA on TRAINING data
    # ---------------------------------------------------------------------

    X_train_selected = selector.fit_transform(
        X_train_scaled,
        y_train
    )


    # ---------------------------------------------------------------------
    # Transform TEST data
    # ---------------------------------------------------------------------

    X_test_selected = selector.transform(
        X_test_scaled
    )


    # ---------------------------------------------------------------------
    # Selected features
    # ---------------------------------------------------------------------

    selected_features = selector.get_selected_features()


    print("\nSelected Features:")
    print(selected_features)


    # ---------------------------------------------------------------------
    # Store everything needed for later evaluation
    # ---------------------------------------------------------------------

    doda_results[k] = {

        "selector": selector,

        "X_train": X_train_selected,

        "X_test": X_test_selected,

        "features": selected_features,

        "raw_math_scores": selector.raw_math_scores_,

        "math_scores": selector.math_scores_,

        "clinical_weights": selector.clinical_weights_,

        "final_scores": selector.final_scores_
    }


print("\n" + "=" * 70)
print("BORUTA + DODA FEATURE SELECTION COMPLETED")
print("=" * 70)
`````

`````python
# CELL 2
# =============================================================================
# INSPECT MATHEMATICAL SCORES
# =============================================================================

k = 10

selector = doda_results[k]["selector"]

# -----------------------------------------------------------------------------
# Raw mathematical scores
# -----------------------------------------------------------------------------

raw_math_scores = pd.DataFrame(
    selector.raw_math_scores_.items(),
    columns=[
        "Feature",
        "Raw Math Score"
    ]
)

# -----------------------------------------------------------------------------
# Normalized mathematical scores
# -----------------------------------------------------------------------------

normalized_math_scores = pd.DataFrame(
    selector.math_scores_.items(),
    columns=[
        "Feature",
        "Normalized Math Score"
    ]
)

# -----------------------------------------------------------------------------
# Combine both
# -----------------------------------------------------------------------------

math_scores = raw_math_scores.merge(
    normalized_math_scores,
    on="Feature"
)

math_scores = math_scores.sort_values(
    by="Normalized Math Score",
    ascending=False
)

print("=" * 70)
print("MATHEMATICAL SCORES")
print("=" * 70)

display(math_scores)
`````

# Interpretation — Boruta + DODA

## 1. Overview

The Boruta baseline and Boruta + DODA experiments were evaluated using four feature-selection sizes:

- Top-5
- Top-10
- Top-15
- Top-20

Each feature subset was evaluated using:

- Logistic Regression
- Random Forest
- XGBoost

The evaluation metrics were:

- Accuracy
- Precision
- Recall
- F1 Score
- ROC-AUC

The purpose of the experiment was to determine whether incorporating clinical knowledge through DODA changes the Boruta-derived feature ranking and whether those changes translate into improved downstream predictive performance.

---

## 2. Boruta Baseline Results

Boruta identified the following features as the highest-ranked features:

### Top-5

1. Age
2. BMI
3. GenHlth
4. HighBP
5. Income

### Top-10

The Top-5 features were followed primarily by:

- PhysHlth
- HighChol
- Education
- MentHlth
- DiffWalk

This indicates that Boruta's data-driven importance ranking placed demographic, metabolic, general-health, and physical-health variables highly.

The strongest Top-5 models were:

| Model | Accuracy | Precision | Recall | F1 | ROC-AUC |
|---|---:|---:|---:|---:|---:|
| Logistic Regression | 0.7043 | 0.3087 | 0.7527 | 0.4378 | 0.7977 |
| Random Forest | 0.7232 | 0.3002 | 0.6083 | 0.4020 | 0.7362 |
| XGBoost | 0.6958 | 0.3045 | 0.7706 | 0.4366 | 0.8007 |

XGBoost achieved the highest recall (0.7706) and ROC-AUC (0.8007) among the Top-5 models.

As the number of features increased, the Random Forest accuracy increased substantially, reaching **0.8424 at Top-20**. However, this was accompanied by a very low recall of **0.1492**, indicating that the high accuracy is largely driven by the majority class.

Therefore, accuracy alone is not an appropriate criterion for selecting the best model in this imbalanced diabetes prediction problem.

---

# 3. Boruta + DODA Results

DODA was subsequently applied to the Boruta ranking using the clinical knowledge weights.

The purpose was not simply to improve prediction, but to modify the data-driven ranking according to clinically informed feature importance.

For Top-5, DODA changed the selected features from:

**Boruta:**

> Age, BMI, GenHlth, HighBP, Income

to:

**Boruta + DODA:**

> BMI, GenHlth, Age, HighBP, PhysHlth

Thus, **PhysHlth replaced Income** in the Top-5 feature set.

This is an important result because it demonstrates that the clinical operator can modify the purely data-driven feature ranking.

# 5. Stroke Feature — Important Observation

An important observation concerns the `Stroke` feature.

In the original Boruta ranking:

```text
Stroke → Rank 19
```

The clinical weighting mechanism strongly increased the importance of Stroke. Consequently:

```text
Stroke → Rank 11
```

Therefore, DODA moved Stroke upward by **8 positions**.

However, despite this substantial increase, Stroke **did not enter the Top-10 feature set**.

This is an important finding rather than a failure of the method.

It demonstrates that the clinical weighting operator does not automatically force a clinically important feature into the selected subset. Instead, the final ranking remains dependent on the interaction between:

1. Data-driven Boruta importance
2. Normalized mathematical importance
3. Clinical weights
4. The selected fusion mechanism

Thus, even a strongly weighted clinical feature can remain below the Top-K threshold when its underlying data-driven importance is relatively low.

This provides evidence that DODA acts as a **ranking adjustment mechanism rather than a hard clinical override**.

# 12. Overall Effect of DODA

The results show three main effects.

### 1. DODA changes the feature ranking

# 13. Interpretation of the Stroke Result

The Stroke result is particularly relevant to evaluating the behavior of DODA.

Although Stroke received a high clinical weight, its original Boruta ranking was very low:

```text
Boruta rank = 19
```

After DODA:

```text
DODA rank = 11
```

Therefore:

```text
Rank improvement = 19 - 11 = 8 positions
```

However, it still remained outside Top-10.

This indicates that the DODA framework is **not functioning as a rule-based feature injection mechanism**.

Instead, clinical knowledge modifies the mathematical ranking while allowing the empirical evidence to remain influential.

This is arguably a desirable property for a clinical feature-selection framework because assigning a high clinical weight should not automatically guarantee selection if the feature has weak empirical support in the dataset.

# 14. Why Top-K Same = False Is Important

The appropriate interpretation is:

> DODA successfully incorporates clinical knowledge into the data-driven feature ranking, resulting in clinically different feature subsets while maintaining comparable predictive performance.

To establish superiority, additional statistical validation would be required, such as:

* repeated stratified cross-validation
* feature-selection stability analysis
* confidence intervals
* statistical comparison of model performance
* clinical validity assessment

# 15. Main Finding

The main finding from the Boruta experiment is **not that DODA dramatically improves accuracy**.

Rather, the experiment demonstrates that:

> **DODA can modify a data-driven Boruta feature ranking using clinically informed weights, promote clinically relevant variables such as Stroke, and produce different Top-K feature subsets while maintaining broadly comparable predictive performance.**

This supports the central motivation of the DODA framework: combining **data-driven feature importance with domain knowledge** rather than relying exclusively on statistical importance.

---

# 16. Research Interpretation

From a research perspective, the Boruta experiment provides evidence for the following hypothesis:

> **A clinically guided feature-selection framework can alter the prioritization of features selected by a purely data-driven method without necessarily compromising downstream predictive performance.**

The diabetes results support this hypothesis.

The strongest evidence is the combination of:

1. **Different feature rankings**
2. **Movement of clinically important features**
3. **Replacement of some data-driven features by clinically prioritized features**
4. **Comparable ROC-AUC and F1 performance**
5. **No substantial deterioration in predictive performance**

However, the current results should be presented as **evidence of feasibility and behavioral effect**, rather than definitive proof that DODA is superior to Boruta.

---

# 17. Important Limitation

The current experiment uses a single train/test split.

Therefore, small differences such as:

```text
ROC-AUC:
0.8056 → 0.8078
```

should not be interpreted as statistically significant improvements.

The next stage should evaluate the methods using repeated or stratified cross-validation and measure:

* Mean ROC-AUC
* Standard deviation
* Mean F1
* Mean recall
* Feature-selection stability
* Jaccard similarity
* Clinical validity

This will determine whether the observed effects of DODA are robust rather than specific to one train/test split.

---

# 18. Final Conclusion

Overall, the Boruta + DODA experiment demonstrates that the DODA framework successfully modifies the feature ranking produced by Boruta.

The most notable example is **Stroke**, which moved from rank **19 to rank 11** after clinical weighting, despite not entering the Top-10 even after an 8-rank improvement.

At the same time, the core high-ranking features such as **BMI, GenHlth, Age and HighBP** remained highly ranked.

Importantly, the resulting feature changes did not cause a substantial deterioration in downstream predictive performance.

Therefore, the experiment supports the interpretation that DODA functions as a **clinical adjustment layer over data-driven feature selection**, rather than as a hard rule-based selector.

The results provide promising evidence that DODA can improve the **clinical alignment and interpretability of feature selection while preserving predictive utility**, but further cross-validation and stability analysis are required before claiming general superiority over Boruta.
