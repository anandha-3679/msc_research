````python
# =============================================================================
# BASELINE MODELS
# =============================================================================

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier


models = {

    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        random_state=42,
        eval_metric="logloss",
        scale_pos_weight=(
            (y_train == 0).sum() /
            (y_train == 1).sum()
        )
    )
}
````

````python
# =============================================================================
# MRMR MODEL EVALUATION
# =============================================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)


mrmr_model_results = []


for k in k_values:

    Xtr = mrmr_results[k]["X_train"]

    Xts = mrmr_results[k]["X_test"]

    features = ", ".join(
        mrmr_results[k]["features"]
    )


    for model_name, model in models.items():

        # -------------------------------------------------------------
        # Train
        # -------------------------------------------------------------

        model.fit(
            Xtr,
            y_train
        )


        # -------------------------------------------------------------
        # Predictions
        # -------------------------------------------------------------

        y_pred = model.predict(
            Xts
        )


        y_prob = model.predict_proba(
            Xts
        )[:, 1]


        # -------------------------------------------------------------
        # Store metrics
        # -------------------------------------------------------------

        mrmr_model_results.append({

            "Method": "MRMR",

            "Top-K": k,

            "Model": model_name,

            "Selected Features": features,

            "Accuracy": accuracy_score(
                y_test,
                y_pred
            ),

            "Precision": precision_score(
                y_test,
                y_pred,
                zero_division=0
            ),

            "Recall": recall_score(
                y_test,
                y_pred,
                zero_division=0
            ),

            "F1 Score": f1_score(
                y_test,
                y_pred,
                zero_division=0
            ),

            "ROC-AUC": roc_auc_score(
                y_test,
                y_prob
            )
        })


# =============================================================================
# CREATE RESULTS DATAFRAME
# =============================================================================

mrmr_results_df = pd.DataFrame(
    mrmr_model_results
)


display(mrmr_results_df)
````

````python
# =============================================================================
# MRMR ADAPTER FOR DODA
# =============================================================================

import numpy as np


class MRMRAdapter:

    def __init__(self, K):
        self.K = K
        self.importance = {}

    def fit(self, X, y):

        # Run MRMR
        selected_features = mrmr_classif(
            X=X,
            y=y,
            K=self.K
        )

        # -------------------------------------------------------------
        # Convert MRMR ranking into importance scores
        # First feature = highest score
        # -------------------------------------------------------------

        n_features = len(selected_features)

        scores = np.zeros(len(X.columns))

        feature_to_index = {
            feature: i
            for i, feature in enumerate(X.columns)
        }

        for rank, feature in enumerate(selected_features):

            index = feature_to_index[feature]

            scores[index] = (
                n_features - rank
            ) / n_features

        # -------------------------------------------------------------
        # Store importance
        # -------------------------------------------------------------

        self.importance = dict(
            zip(
                X.columns,
                scores
            )
        )

        return self

    def get_importance(self):

        return self.importance

    def get_name(self):

        return "MRMR"
````

````python
# =============================================================================
# MRMR + DODA FEATURE SELECTION
# =============================================================================

import numpy as np
import pandas as pd


# -------------------------------------------------------------------------
# Top-K values
# -------------------------------------------------------------------------

k_values = [5, 10, 15, 20]


# -------------------------------------------------------------------------
# Clinical knowledge provider
# -------------------------------------------------------------------------

provider = JSONProvider(
    "clinical_weights.json"
)


# -------------------------------------------------------------------------
# Fusion method
# -------------------------------------------------------------------------

fusion = HadamardFusion()


# -------------------------------------------------------------------------
# MRMR Adapter
# -------------------------------------------------------------------------

class MRMRAdapter:

    def __init__(self, ranked_features):

        self.ranked_features = ranked_features
        self.importance = {}


    def fit(self, X, y):

        n_features = len(self.ranked_features)

        # Convert MRMR ranking into decreasing scores
        # Rank 1 = highest score
        # Rank n = lowest score

        for rank, feature in enumerate(
            self.ranked_features,
            start=1
        ):

            self.importance[feature] = (
                n_features - rank + 1
            ) / n_features

        return self


    def get_importance(self):

        return self.importance


    def get_name(self):

        return "MRMR"


# -------------------------------------------------------------------------
# Store results
# -------------------------------------------------------------------------

mrmr_doda_results = {}


# =============================================================================
# RUN MRMR + DODA
# =============================================================================

for k in k_values:

    print("\n" + "=" * 70)
    print(f"MRMR + DODA — TOP {k} FEATURES")
    print("=" * 70)


    # ---------------------------------------------------------------------
    # Create MRMR adapter
    # ---------------------------------------------------------------------

    mrmr_adapter = MRMRAdapter(
        mrmr_ranked_features
    )


    # ---------------------------------------------------------------------
    # DODA selector
    # ---------------------------------------------------------------------

    selector = DODASelector(

        operators=[
            mrmr_adapter
        ],

        provider=provider,

        fusion=fusion,

        top_k=k
    )


    # ---------------------------------------------------------------------
    # Fit DODA on training data
    # ---------------------------------------------------------------------

    X_train_selected = selector.fit_transform(
        X_train_scaled,
        y_train
    )


    # ---------------------------------------------------------------------
    # Transform test data
    # ---------------------------------------------------------------------

    X_test_selected = selector.transform(
        X_test_scaled
    )


    # ---------------------------------------------------------------------
    # Selected features
    # ---------------------------------------------------------------------

    selected_features = (
        selector.get_selected_features()
    )


    print("\nSelected Features:")
    print(selected_features)


    # ---------------------------------------------------------------------
    # Store results
    # ---------------------------------------------------------------------

    mrmr_doda_results[k] = {

        "selector": selector,

        "X_train": X_train_selected,

        "X_test": X_test_selected,

        "features": selected_features,

        "raw_math_scores":
            selector.raw_math_scores_,

        "math_scores":
            selector.math_scores_,

        "clinical_weights":
            selector.clinical_weights_,

        "final_scores":
            selector.final_scores_
    }


print("\n" + "=" * 70)
print("MRMR + DODA FEATURE SELECTION COMPLETED")
print("=" * 70)
````

# MRMR + DODA Feature Selection — Interpretation

## 1. Overview

MRMR (Minimum Redundancy Maximum Relevance) was used as the mathematical feature-selection operator within the DODA framework.

The MRMR ranking was first generated from the training data. DODA then integrated the MRMR scores with domain-specific clinical weights using the Hadamard fusion mechanism:

\[
S_{DODA}(f) = S_{MRMR}(f) \times W_{clinical}(f)
\]

where:

- \(S_{MRMR}(f)\) = normalized MRMR score of feature \(f\)
- \(W_{clinical}(f)\) = clinical importance weight assigned to feature \(f\)
- \(S_{DODA}(f)\) = final DODA score

Top-K features were then selected for \(K = 5, 10, 15,\) and \(20\).

---

## 2. MRMR Ranking Before Clinical Adjustment

The MRMR operator produced the following ranking:

1. GenHlth
2. HighBP
3. BMI
4. HighChol
5. DiffWalk
6. HeartDiseaseorAttack
7. Age
8. Income
9. PhysHlth

11. CholCheck
12. Stroke

14. Education
15. Veggies
16. Smoker
17. MentHlth
18. Sex
19. AnyHealthcare
20. Fruits
21. NoDocbcCost

This represents the purely mathematical ranking produced by MRMR without incorporating clinical knowledge.

---

## 3. Effect of Clinical Knowledge

DODA modifies the MRMR scores using predefined clinical weights.

Several clinically important features received high weights:

- GenHlth → 1.0
- HighBP → 1.0
- BMI → 1.0
- HeartDiseaseorAttack → 1.0
- Stroke → 1.0
- HighChol → 0.9
- DiffWalk → 0.9
- Age → 0.9
- PhysHlth → 0.8

In contrast, features such as Income, HvyAlcoholConsump, Fruits, and NoDocbcCost received lower clinical weights.

For example:

- Income: 0.6667 → 0.20 after clinical adjustment
- HvyAlcoholConsump: 0.5714 → 0.1143
- Stroke: 0.4762 → 0.4762

This demonstrates that DODA does not simply preserve the mathematical ranking. It modifies feature priority according to domain knowledge.

---

## 4. Important Ranking Changes

The most important effect is observed around the middle of the MRMR ranking.

### Stroke

Stroke had:

- MRMR rank: **12**
- MRMR normalized score: **0.4762**
- Clinical weight: **1.0**
- DODA score: **0.4762**

Because Stroke received the maximum clinical weight, its score was completely preserved.

Consequently, Stroke moved from **rank 12 to rank 9** in the DODA ranking.

This is an important result because Stroke is clinically relevant to cardiovascular/diabetes-related risk, but its purely mathematical MRMR score was not sufficiently high to place it in the Top-10.

### Income

Income had:

- MRMR rank: **8**
- MRMR score: **0.6667**
- Clinical weight: **0.3**
- DODA score: **0.20**

Therefore, Income was strongly down-weighted by the clinical knowledge layer.

### HvyAlcoholConsump

HvyAlcoholConsump had:

- MRMR rank: **10**
- MRMR score: **0.5714**
- Clinical weight: **0.2**
- DODA score: **0.1143**

It therefore received substantially lower priority after clinical adjustment.

# MRMR + DODA Results Interpretation

## 1. Overall Performance

The MRMR-based feature selection was evaluated using Logistic Regression, Random Forest, and XGBoost for Top-K feature subsets of 5, 10, 15, and 20 features.

The results show that **MRMR + DODA generally maintains or slightly improves the predictive performance of MRMR alone for Logistic Regression and XGBoost**, while Random Forest shows a different behavior.

---

## 2. MRMR Baseline

### Top 5 Features

The MRMR baseline selected:

- GenHlth
- HighBP
- BMI
- HighChol
- DiffWalk

The best performance among the three models was obtained by **Logistic Regression in terms of F1 score (0.4334)**, while XGBoost achieved the highest Recall (0.7645) and ROC-AUC (0.7946).

This indicates that the Top-5 MRMR features contain substantial predictive information, but the model still faces the challenge of class imbalance.

### Top 10–20 Features

Increasing the number of features generally improved the ROC-AUC of Logistic Regression and XGBoost.

For example, XGBoost achieved:

- Top 5: ROC-AUC = **0.7946**
- Top 10: ROC-AUC = **0.8083**
- Top 15: ROC-AUC = **0.8103**
- Top 20: ROC-AUC = **0.8118**

This suggests that additional features provide useful complementary information beyond the initial five MRMR-selected features.

Random Forest showed a different pattern. Its accuracy increased considerably with larger feature sets, but its Recall and F1 score decreased. Therefore, the high accuracy should not be interpreted as better predictive performance because the dataset is imbalanced.

---

## 3. MRMR + DODA

DODA incorporates clinical knowledge into the mathematical ranking produced by MRMR.

The MRMR ranking initially placed `Income` above `PhysHlth`, `HvyAlcoholConsump`, and `Stroke`.

After applying the clinical knowledge layer:

- `HeartDiseaseorAttack` moved upward.
- `Stroke` moved substantially upward.
- `PhysHlth` moved upward.
- `Income` was down-weighted because of its lower clinical relevance.
- Several lifestyle and socioeconomic variables received lower final scores.

For example, the MRMR ranking was:

`GenHlth → HighBP → BMI → HighChol → DiffWalk → HeartDiseaseorAttack → Age → Income → PhysHlth → HvyAlcoholConsump → CholCheck → Stroke`

After DODA:

`GenHlth → HighBP → BMI → HighChol → HeartDiseaseorAttack → DiffWalk → Age → PhysHlth → Stroke → CholCheck → Income → ...`

This demonstrates that DODA does not simply reproduce the mathematical ranking. Instead, it **reorders features according to both statistical evidence and clinical knowledge**.

---

## 4. Effect on Top-K Feature Selection

### Top 5

**MRMR:**

`GenHlth, HighBP, BMI, HighChol, DiffWalk`

**MRMR + DODA:**

`GenHlth, HighBP, BMI, HighChol, HeartDiseaseorAttack`

DODA replaces `DiffWalk` with `HeartDiseaseorAttack`.

This is an important result because both features have strong clinical relevance, but DODA prioritizes the cardiovascular history variable based on the assigned clinical knowledge.

---

### Top 10

**MRMR + DODA:**

The major changes are:

- `Income` moves down.

This demonstrates the intended behavior of the clinical adjustment layer.

The results are particularly encouraging because DODA changes the feature ranking while maintaining almost identical predictive performance.

This suggests that the clinically prioritized features can replace some statistically ranked features **without sacrificing predictive discrimination**.

### Random Forest

Random Forest behaves differently.

Its accuracy increases substantially at Top-15 and Top-20, but Recall and F1 decrease.

For example:

- MRMR Top-20: Accuracy = **0.8436**, Recall = **0.1571**, F1 = **0.2351**
- MRMR + DODA Top-20: Accuracy = **0.8423**, Recall = **0.1510**, F1 = **0.2265**

Therefore, the higher accuracy is mainly associated with predicting the majority class and should not be considered evidence of improved clinical performance.

For this imbalanced dataset, **Recall, F1 score, and ROC-AUC are more informative than accuracy alone**.

---

## 6. Key Finding

The most important finding from the MRMR experiment is that **DODA changes the feature ranking substantially while preserving predictive performance**.

The mathematical MRMR ranking favors features such as `Income` and `HvyAlcoholConsump` relatively highly. After incorporating clinical knowledge, DODA promotes clinically meaningful variables such as `HeartDiseaseorAttack`, `Stroke`, and `PhysHlth`.

Despite these changes, Logistic Regression and XGBoost maintain almost the same ROC-AUC values.

Therefore, the results provide preliminary evidence that:

> **DODA can introduce clinical knowledge into feature selection without substantially reducing predictive performance.**

This is an important result for the proposed clinically guided feature-selection framework.

---

## 7. Important Observation About Stroke

`Stroke` provides a particularly useful demonstration of the DODA mechanism.

Under MRMR, `Stroke` was ranked **12th**, which excluded it from the Top-10 feature set.

After applying DODA, its clinical weight of **1.0** increased its final score and moved it to **9th position**, allowing it to enter the Top-10 feature set.

This demonstrates that DODA can prevent clinically important variables from being excluded solely because their statistical ranking is relatively low.

---

## 8. Conclusion

Overall, the MRMR experiment supports the proposed DODA framework.

The results demonstrate three important behaviors:

1. **MRMR identifies statistically relevant and complementary features.**
2. **DODA modifies the ranking according to predefined clinical knowledge.**
3. **The clinical adjustment does not substantially reduce predictive performance.**

The similarity in ROC-AUC between MRMR and MRMR + DODA, particularly for Logistic Regression and XGBoost, suggests that the clinically guided feature sets retain most of the predictive information of the original MRMR selections.

However, these results should be considered **preliminary evidence**. Further validation using repeated cross-validation, feature-selection stability analysis, and comparison across multiple datasets is required before concluding that DODA consistently outperforms conventional feature-selection methods.
